## <a> Checking for Normality: Quantile-Quantile (QQ) Plots </a>
---

A <font color="dodgerblue">**Quantile-Quantile plot** (qq-plot)</font>  *compares a variable's quantiles to those of a normal distribution*.
- Its *horizontal* axis contains the quantiles of a normal distribution (what you would expect if the data was normal).
- Its *vertical* axis contains the quantiles from your sample.
- It also has a diagonal <font color="violet">*reference line*</font> that shows where all the points would fall if the variable was normally distributed.

How to interpret a qq-plot:

- If the points fall roughly **along the reference line**, the data is *approximately normal*.
- If the points **curve or deviate from the reference line** (particularly between the 1st and 3rd quantiles), the data is *not approximately normal*.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/qq_plot_examples.png?raw=true" width=900/>
</figure>

### When to use it

When you want a quick visual check for normality before running parametric tests.

### Be aware of

"Perfect" straight lines are rare, so focus on *major* curves or outliers. Remember that large samples can make minor deviations look dramatic.

### In R

To make the qq-plot itself, use <font color="tomato">`qqnorm(<variable>)`</font>, then add the reference line using <font color="tomato">`qqline(<variable>)`</font>.

## <a> Testing for Normality: Shapiro-Wilk Test </a>
---

The <font color="dodgerblue">**Shapiro-Wilk Test**</font> asks "could this sample reasonably have come from a normal distribution?" by basically quantifying what we're seeing in the QQ plots. It does it by calculating the following ratio:

$$W = \frac{expected \,\, variance \,\, if \,\, data \,\, is \,\, normal}{sample\,\, variance}$$

The closer the ratio is to 1, the more closely the sample resembles a normal distribution.

The test statistic is commonly named $W$. Converting the ratio to a $p$-value requires complex polynomial approximations and transformations, so there is no table to refer to here. We must rely on technology.

- $H_0$: the data comes from a normal distribution.
- $H_1$: the data does NOT come from a normal distribution.

### When to use it

When you want a formal, $p$-value based test of normality.

### Be aware of

With large samples, this test can flag small deviations as significant. With a small sample size, this test has low power. With a large sample size, it becomes overly sensitive to minor deviations. A good "sweet spot" for sample size here is $n \approx 20-100$.

### In R

To run a Shapiro-Wilk test, use <font color="tomato">`shapiro.test(<variable>)`</font>.

# <a> Non-normal group comparison: Wilcoxon Rank-Sum Test</a>
---

Also referred to as the *Mann-Whitney U* test, it's used to compare two groups ("are the groups different?") similar to an independent / Welch's $t$ test.

If a group is not normally distributed, then the mean is no longer a good value to compare. So what do we do instead? Short answer: we rank all observations in order of largest to smallest, then add the ranks for each group. This gives us how separated the groups' ranks are, and therefore how different the values are.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/mann_whitney_u_test.png?raw=true" width=400/>
</figure>

- $H_0$: the two groups' values are not significantly higher or lower than each other.
- $H_1$: one group has different values than the other

Like the Shapiro-Wilk test, the test statistic is also commonly named $W$ (though calculated differently) and is similarly hard to calculate, so we must rely on technology once again.

### When to use it

When at least one of the groups is not approximately normal or when outliers make comparing the means unreliable. It's also useful if your values are ordinal instead of nominal.

### Be aware of

Since it compares central tendencies instead of means, it's not *exactly* an equivalent of the independent $t$ test.

### In R

To run a two-sided Wilcoxon test, use <font color="tomato">`wilcox.test(<first_vector>, <second_vector>, alternative="two.sided")`</font>.

# <a>Comparing Means with Unequal Variances: Welch's $t$-test </a>
---

The <font color="dodgerblue">**Welch's $t$ test**</font> has the same process, hypothesis, and test statistic as the independent $t$ test. The difference is how the degrees of freedom is handled: instead of using a whole number of $n_1 + n_2 -2$, it weighs the variances by each group's sample size contribution.

$\large df \approx \frac{(\frac{s^2_1}{n_1} + \frac{s^2_2}{n_2})^2}{\frac{(s^2_1/n_1)^2}{n_1 - 1} + \frac{(s^2_2/n_2)^2}{n_2 - 1}}$

### When to use it

When your distributions are both approximately normal but don't have equal variances.

### Be aware of

Since the weighted variance can have a decimal value, a distribution table won't have the critical or $p$-values directly.

### In R

To run a two-sided Welch's $t$ test, use <font color="tomato">`t.test(<first_vector>, <second_vector>, var.equal=FALSE, alternative="two.sided")`</font>. If you run `t.test()` without telling it whether the variances are equal, it will default to FALSE and run a Welch's $t$ test.
