# <a name="08normal-intro">Normal Distributions</a>

---

There are many situations where data tends to be distributed
*symmetrically around a central value* with no skew to the left or right.
For example, if we randomly select an adult from the population:

-   A person is most likely to be about average height.
-   Otherwise, the person selected is equally likely to be either shorter or taller than average.
-   The person is very unlikely to have an extreme height (tall or short). The more extreme, the less likely.
-   **Therefore the distribution of heights is approximately symmetric and bell shaped.**

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/bell_curve.webp?raw=true" width=400/>
</figure>

Normal distributions are very useful and probably the most important distribution in statistics!

<br>

## <a> Properties </a>

<font color="dodgerblue">**Normal distributions**</font> arise in
many settings: heights of people, size of items produced by machines,
and most importantly in statistics, *data sets resulting from many
independent random events*. The shape of a normal distribution is
determined by two parameters:

-   The <font color="dodgerblue">**mean**</font>, $\color{dodgerblue}{\mu}$, is the center of the distribution.
-   The <font color="dodgerblue">**standard deviation**</font> $\color{dodgerblue}{\sigma}$, measures the spread or variability of the data.

Notation: if $X$ is normally distributed with mean $\mu$ and standard deviation $\sigma$, then we write $\color{dodgerblue}{\mathbf{X \sim N(\mu, \sigma)}}$.

This nice symmetry makes normal distributions very easy to work with and give us other useful features, like the **empirical rule** and **standardization**, which we'll learn about a bit later.

<br>

## <a> The formal maths stuff </a>

If $X$ is <font color="dodgerblue">**normally distributed**</font>
with mean $\mu$ and standard deviation $\sigma$, we write
$\color{dodgerblue}{X \sim N( \mu ,\sigma)}$. The formula for the
probability density is called the Gaussian and is given below.

$$f(x; \mu, \sigma) = \dfrac{1}{\sigma\sqrt{2\pi}} e^{-\frac{1}{2} \left( \frac{x-\mu}{\sigma} \right)^2}$$

-   The expected value is $E(X) = \mu$.
-   The variance is $\mbox{Var} (X) = \sigma^2$.

## <a> R Functions </a>

-   Use `dnorm(x, mu, sigma)` to calculate the PDF $f(x)$.
    -   <font color="tomato">Recall for continuous distributions,
        $f(x)$ gives the **height** of the pdf.</font>
    -   <font color="tomato">Probabilities are areas under $f(x)$,not the heights.</font>
    -   <font color="tomato">**DO NOT use `dnorm()` to compute
        probabilities.**</font>
-   <font color="mediumseagreen">**Be sure to use `pnorm(x, mu, sigma)` to calculate the CDF $P(X \leq x)$, the area to the left of $X=x$.**</font>



## <a name="08append-normal">PDF and CDF of a Normal Distribution</a>

---

Here is a plotting example of a normal distribution whose mean is 40 and standard deviation is 5.

In notation, that is ${\mathbf{X \sim N(40, 5)}}$




```r
options(repr.plot.width = 10, repr.plot.height = 6) # adjust plot window size
par(mfrow=c(1,2)) # puts the plots next to each other

x <- seq(15, 65, length=1000)

mean <- 40
sd <- 5

plot(x, dnorm(x, mean, sd), type="l", col = "red", xlab="x value",
  ylab="Density", main="pdf for N(40,5)")

plot(x, pnorm(x, mean, sd), type="l", col = "blue", xlab="x value",
  ylab="Density", main="cdf for N(40,5)")
```

## <a name="08append-normal">How the standard deviation $\sigma$ affects the distribution shape</a>

---

Since the standard deviation $\sigma$ is a measure of how spread out the values in a distribution are, it can affect the shape of the distribution.

- If $\sigma$ is *small*, it means that the values are close to the mean, making the distribution **tall and skinny**.

- If $\sigma$ is *large*, it means that the values are far away from the mean, making the distribution **shorter and wider**.

```r
options(repr.plot.width = 6, repr.plot.height = 6) # adjust plot window size

# run code cell to compare distributions
# with different standard deviations
# nothing to edit in this code cell

x <- seq(-10, 10, length=1000)
stan <- dnorm(x)

std <- c(1, 2, 4)
colors <- c("black", "blue", "red")
labels <- c("sigma=1", "sigma=2", "sigma=4")

plot(x, stan, type="l", lty=2, xlab="x value",
  ylab="Density", main="Comparison of St Dev")

for (i in 1:3){
  lines(x, dnorm(x, 0, std[i]), lwd=2, col=colors[i])
}

legend("topright", inset=.05, title="Distributions",
  labels, lwd=2, lty=1, col=colors)
```

## <a name="08append-normal">How the mean $\mu$ affects the distribution location</a>

---

Since normal distributions center around the mean, its value can shift the distribution along the x axis.

*Note: the mean does not affect the shape of the distribution, just its location on the horizontal axis.*

```r
# run code cell to compare distributions
# with different means
# nothing to edit in this code cell

x <- seq(-10, 10, length=100)
stan <- dnorm(x)
mu <- c(0, 5, -5)
colors <- c("black", "blue", "red")
labels <- c("mu=0", "mu=5", "mu=-5")

plot(x, stan, type="l", lty=2, xlab="x value",
  ylab="Density", main="Comparison of Means")

for (i in 1:3){
  lines(x, dnorm(x, mu[i], 1), lwd=2, col=colors[i])
}

legend("topright", inset=.05, title="Distributions",
  labels, lwd=2, lty=1, col=colors)
```

# <a name="08empirical">The 68%-95%-99.7% Empirical Rule</a>

---

This is one of the advantages of normal distributions being symmetrical.

The <font color="dodgerblue">**empirical rule**</font> for normal
distributions states that approximately:

-   68% of all values fall within *one* standard deviation (both above and
    below) from the mean.
-   95% of all values fall within *two* standard deviations of the mean.
-   99.7% of all values fall within *three* standard deviations of the
    mean.

# <a name="19intro">The Central Limit Theorem (CLT)</a>

---

In the beginning of the course, we talked about sample vs population:

| **Example**                 | **Population** (everything we want to study)                         | **Sample** (a subset of the population we want to study)                                               |
| --------------------------- | -------------------------------------- | --------------------------------------------------------- |
| **Type 1 Diabetes**      | All female people with type 1 diabetes | 300 female UCHealth patients with type 1 diabetes         |
| **Cancerous Skin Cells** | All cancerous skin cells in a patient  | A culture of 4,000 cancerous skin cells from that patient |

It's not always practical or possible to measure every single item/patient/etc in a population, so we use a subset of it (a sample) to try to learn information about the population we care about.

<br>

Keeping that in mind, let's talk about the **Central Limit Theorem**.

In the simplest terms, the CLT says that <font color="dodgerblue"> **as your sample size gets bigger, you sample distribution will get closer and closer to a normal distribution**, *regardless of what underlying distribution your sample originally came from*. </font>

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/clt2.gif?raw=true" width=400/>
</figure>

This is very helpful because when we collect sample data, we don't always know what the underlying distribution is - **but**, if we collect enough of it, we can use a normal distribution and its very useful properties to analyze and learn from our sample.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/clt1.jpg?raw=true" width=400/>
</figure>

This also means that <font color="mediumseagreen">**the more sample data we collect (as $n$ increases), the *mean of our sample* ($\bar{x}$) is more and more likely to be close to the *mean of the population* ($\mu$)**</font>, and the sample distribution becomes more and more concentrated around $\mu$. This phenomenon is called the <font color="dodgerblue">Law of Large Numbers</font>.

This is useful when we want to try to approximate the population mean $\mu$.
