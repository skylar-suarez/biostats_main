## <a> Variables </a>

We often would like to investigate whether one variable is associated to another. Researchers carry out studies to understand the conditions and causes of certain outcomes.

If we are using one variable to help us understand or predict the values (or category) of another variable:

- We call the first variable the <font color="dodgerblue">**independent** or  **predictor**</font> variable.
- The second the <font color="dodgerblue">**dependent** or  **response**</font> variable.
  - Different categories of a predictor variable are called <font color="dodgerblue">**treatments** or **levels**</font>.
- A variable that can influence the relationship between the independent and dependent variables is called a <font color="dodgerblue">**confounding variable**</font>.
- A variable that we hold constant or adjust to isolate the effect we want to study is a <font color="dodgerblue">**control variable**</font>. Controls help us minimize or eliminate influence from confounders.

## <a name="01expriment">Experiments and Observational Studies</a>

In an <font color="dodgerblue">**observational study**</font>, the researcher *does not actively control* the assignment of individuals to different treatments or levels of a predictor variable.

- If the treatment groups are chosen by the individuals being studied, the samples in each treatment group are likely to differ in some meaningful way other than just the treatment.

<font color="tomato">**There are almost always confounding variables in observational studies. Thus observational studies can almost never be used to establish *causation*.**</font>

In an <font color="dodgerblue">**experiment**</font>, the researcher *controls the assignment* of individuals to different treatments or levels of an independent variable.

- In a <font color="dodgerblue">**randomized experiment**</font>, the predictor variable for each individual is determined randomly, before the experiment is carried out. If treatment groups are randomly determined, they should be fairly similar in every way except for the treatment itself.

<font color="mediumseagreen">**When properly designed, randomized experiments can show a predictor variable *causes* a change in the response variable.**</font>

Not every question can be directly addressed with an experiment due to ethical reasons (e.g., we can't give human beings cancer to study the effects). In such cases, we would use observational studies to investigate *association*.

<br>

Here's a quick summary:


| Observational Studies                    | Randomized Experiments                           |
| ------------------------------------- | ---------------------------------------- |
| Easier, faster, cheaper | Expensive, complex, time-consuming |
| *Less* control over how data is collected | *More* control over how data is collected |
| Can study exposures that would be unethical to assign | Some questions cannot be studied for ethical reasons |
| Cannot establish causal links due to confounders | Stronger evidence for causality by removing confounders |

# <a name="01sampling">Random Sampling Methods</a>

Since drawing a sample that resembles the population in every way
(except smaller in number) is critical for drawing valid conclusions, how we pick samples is crucial.

<font color="dodgerblue">**Sampling bias**</font> occurs when the
method of selecting a sample causes the sample to differ from the population in some relevant way. When possible, collecting samples *randomly* without human interference is the best and therefore preferred method. <font color="dodgerblue">**Randomly selecting samples is the best way to avoid bias.**</font> Below are some common methods for selecting random samples and avoiding bias.

### Simple Random Sample

When selecting a <font color="dodgerblue">**simple random sample**</font>, all individuals are equally likely to be selected.

-   Example: Assign 12 participants in an experiment a number 1 to 12. Randomly select 4 numbers and assign those individuals to treatment group A.

<figure>
<a title="Dan Kernler, CC BY-SA 4.0 &lt;https://creativecommons.org/licenses/by-sa/4.0&gt;, via Wikimedia Commons" href="https://commons.wikimedia.org/wiki/File:Simple_random_sampling.PNG"><img width="256" alt="Simple random sampling" src="https://upload.wikimedia.org/wikipedia/commons/b/bf/Simple_random_sampling.PNG"></a>
</figure>

### Stratified Sample

When selecting a <font color="dodgerblue">**stratified
sample**</font>, the population is subdivided into groups based on some meaningful characteristic *prior to assignment*.

-   Example: Divide all voters into categories based on political party. Then randomly sample individuals from each political party so the overall sample has the same percentage of voters in each party as the population of all voters.

# <a name="02structure">The Structure of Data</a>
---

<font color="dodgerblue">**Data frames**</font> are
two-dimensional data objects and are the **fundamental** data structure used by most of R’s libraries of functions and data sets.

**Tabular data is <font color="dodgerblue">**tidy**</font> if each row corresponds to a different observation and each column corresponds to a different variable.**

Each column of a data frame is a <font color="dodgerblue">**variable**</font> (stored as a **vector**) of possibly different data types.

-  If a variable is measured or counted by a number, it is called a <font color="dodgerblue">**quantitative**</font> or <font color="dodgerblue">**numerical**</font> variable.
  -  Quantitative variables may be <font color="dodgerblue">**discrete**</font> (integers) or <font color="dodgerblue">**continuous**</font> (decimals).
-  If a variable groups observations into different categories or rankings, it is a <font color="dodgerblue">**qualitative**</font> or <font color="dodgerblue">**categorical**</font> variable.
  -   The different categories of a qualitative variable are called <font color="dodgerblue">**levels**</font> or <font color="dodgerblue">**classes**</font>.
    -   Levels are typically labeled with descriptive character strings or integers.
    -   Levels may or may not have an ordering.
  - An <font color="dodgerblue">**ordinal variable**</font> is when the levels of a categorical variable have a specified order (e.g., 0 means no heart failure, 1 means mild heart failure, 2 means severe heart failure).

---

In the command `typeof(storms$status)`, notice we refer to just the `status` variable of the `storms` data frame. <font color="dodgerblue">**A variable from a data frame may be extracted using `$` and then specifying the name of the desired variable**.</font>

-   Followed by a dollar sign `$`.
-   Then indicate the name of the variable, for example `wind`.

# <a name="02data-types">Data Types in R</a>
---

R has 5 basic data types:

1.  <font color="dodgerblue">**character**</font>: collections of characters. E.g., `"a"`, `"hello world!"`
2.  <font color="dodgerblue">**double**</font>: decimal numbers. e.g., `1.2`, `1.0`
3.  <font color="dodgerblue">**integer**</font>: whole numbers. In R, you must add `L` to the end of a number to specify it as an integer. E.g., `1L` is an integer but `1` is a double.
4.  <font color="dodgerblue">**logical**</font>: Boolean values, `TRUE` and `FALSE`
5.  <font color="dodgerblue">**complex**</font>: complex numbers. E.g., `1+3i`




## <a name="02typeof">Checking Data Types Using `typeof()` and `is.datatype()`</a>
---

-   The `typeof()` function returns the R internal type or storage mode
    of any object.
-   The `is.numeric()` function returns a boolean (true or false) depending on if the object is numerical or not.
-   The `is.character()`, `is.double()`, `is.integer()`, etc functions return a boolean depending on if the object is a character, double, integer, etc or not.

## <a name="02decimal-integer">Converting from one data type to another</a>
---

From the summary of the `storms` data set we first found above, we see
that the variables `year` and `month` are being stored as `double`. These variables actually are integer values.

We can convert another variable of one format into another format using
`as.[new_datatype]()`

-   For example, to convert to year to `integer`, we use `as.integer(storms$year)`.
-   To convert a data type to character, we can use `as.character(x)`.
-   To convert to a decimal (`double`), we can use `as.numeric(x)`

```r
typeof(storms$year)
typeof(storms$month)
storms$year <- as.integer(storms$year)
storms$month <- as.integer(storms$month)
typeof(storms$year)
typeof(storms$month)
```

## <a name="02factor">Converting to Categorical Data with `factor()`</a>
---

Sometimes we think a variable is one data type, but it is actually being stored (and thus interpreted by R) as a different data type. One common issue is categorical data is stored as characters or integers. We would like observations with the same values to be group together.

-   The `status` variable in `storms` is being properly stored as a `factor`.
-   The `category` variable in `storms` is being stored as a `numeric` since the classes are integers.



```r
summary(storms$category)
```

The summary of `category` computes statistics such as mean and median. <font color="dodgerblue">**Typically with categorical data, we prefer to count how many observations are in each class of the variable.**</font>

-   In the code cell below, we convert `category` to a factor, and then observe the resulting summary.

```r
storms$category <- factor(storms$category)
summary(storms$category)
```
