# <a> Effect Size </a>
---

So far, we've only used $p$-values to determine if there's a statistically significant difference between two groups in some way. They are very helpful, but they only tell part of the story - what they don't tell you is the magnitude of the difference. <font color="dodgerblue">**Effect size**</font> is a way to <font color="dodgerblue">quantify the difference between two groups</font> (e.g., how much of an impact a treatment had).

Here's an example that illustrates it well:

> *Imagine a recently published study of 10,000 adults shows that reducing daily sodium intake by 500 mg leads to significant reduction in average systolic blood pressure.*

Sounds great, right? Lower blood pressure would greatly improve people's health. We might even consider changing the nationwide dietary guidelines to align with this new finding!

> *Reading further into the publication, we find out that the decrease in systolic blood pressure between tthe control and experimental groups is only 1.5 mmHg with a $p$-value < 0.01.*

The difference was fairly significant, but a reduction of 1.5 mmHg is rather small, isn't it? It isn't very likely to have a meaningful impact on individual health. If achieving this reduction requires drastic diet changes, added food regulations, and substantial costs to food production, policy makers may conclude that the benefit does not outweigh the costs here.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/eff_size.png?raw=true" width=400/>
</figure>

The key takeaway here is that, in practice, <font color="mediumseagreen">**statistical significance ("does an effect exist?") alone does not give enough context for our experiment results, but effect sizes give that context and help us assess the real-world relevance of our findings**</font>.

In general, effect size is just the ratio of difference divided by variability: $\frac{difference}{standard\,deviation}$

There are several ways to calculate this, but we will cover the most used: <font color="dodgerblue">Cohen's d</font>.

## <a> Cohen's $d$ </a>
---

This is widely used when comparing the means of two groups. It does assume that the two groups have variance homogeneity (roughly equal variance). The formula is:

$$\color{dodgerblue}{d = \frac{\bar{x}_1 - \bar{x}_2}{\sqrt{\frac{s_1^2(n_1 - 1) + s_2^2(n_2 -1)}{n_1 + n_2 - 2}}}}$$

If the denominator looks familiar, it should. It's the pooled standard deviation of the two groups $s_{p}$.

Note that this Cohen's d calculation may be negative or positive based on which group you designate to be 1 or 2:
- Positive (+) if group 1 had the larger mean.
- Negative (-) if group 2 had the larger mean.
- For pure magnitude of difference without directionality, take the absolute value of the result.

Base R does not have a built-in function for this, but the `effsize` package has a `cohensD()` function, if you wish to explore that.

### Interpretation

<font color="dodgerblue">The result expresses how many standard deviations the two groups differ by</font>. Interpreting Cohen's d can be tricky because it does not have a neat closed range like correlation does. A difference of 0.2 standard deviations might be considered big in some contexts. However, there is a rough guide, made by Cohen himself:

| Cohen's d | Interpretation |
|---|---|
| 0.2 | small difference |
| 0.5 | moderate difference |
| 0.8 | large difference |
| 1.2 | very large difference |

Once calculated, write an interpretation using the full context from the question, including Cohen's d. For example:

> There is a large difference in average blood pressure between the control and experimental groups ($d = 0.87$).

<font color="mediumseagreen">**If you also conducted a hypothesis test, you should report both results in the same statement.**</font> For example:

> There is a statistically significant difference in average blood pressure between the control and experimental groups, with a large effect size ($p = 0.015, d = 0.87$).

## <a> Sampling Variability </a>
---

So far we've only run hypothesis tests with one sample. Is that enough to truly detect whether a difference or effect exists?

When we take a sample from a population, that sample may not aways be a faithful representation of the *entire* population. Each time we resample, the sample mean will be different - sometimes by a little, sometimes by a lot. We discussed this when we did confidence intervals. This natural variation is called <font color="dodgerblue">**sampling variability**</font>. We measure this variability using the <font color="dodgerblue">**standard error (SE)**</font>, which we also covered during confidence intervals.

If you measure the systolic blood pressure of 100 people:
- You may get 110, 125, 138, 150, 170... mmHg
- This *spread among individual people* is measured by the *standard deviation*.

If you run 100 studies of 100 people each:
- Each study has its own average 129.8, 130.6, 131.2, 129.4, 130.4... mmHg
- This *spread among averages* is the *standard error*, which is just the standard deviation of the averages.
- The average of all 100 sample averages is called the <font color="dodgerblue">**grand sample mean**</font>.

> Generally speaking, <font color="mediumseagreen">**variability gets smaller as your sample size increases**</font>. We covered this briefly when we talked about the central limit theorem and the law of large numbers. Basically, the more observations in your sample, the more confident you can be that you're converging toward the true average value.

# <a> Error Types </a>
---

As we know by now, the result of a hypothesis test has one of two possibilities:

-   If <font color="mediumseagreen">**p-value $\leq \alpha$, we reject $H_0$**</font>.
-   If <font color="tomato">**p-value $> \alpha$, we fail to reject $H_0$**</font>. The test is inconclusive.

Back in week 3, we discussed the possible results of screening tests: false positive, false negative, true positive, true negative. Hypothesis tests have very similar possible outcomes. Here, we'll focus on the incorrect ones:

- A <font color="dodgerblue">**type I error**</font> occurs if we **incorrectly reject $H_0$ when it is true**. This is also known as a <font color="dodgerblue">**false positive**</font>.
    - The significance level of a hypothesis test <font color="orange"> $\alpha$</font> is the <font color="orange">probability of a Type I error</font>.
- A <font color="dodgerblue">**type II error**</font> occurs if we **incorrectly fail to reject $H_0$ when it is false**. This is also known as a <font color="dodgerblue">**false negative**</font>.
    - The <font color="orange">probability of a Type II error</font> is denoted as <font color="orange">$\beta$</font>.

For example, when a jury is deciding a case in court, the hypotheses would be:

- $H_0$: The accused person is innocent (we assume the person is innocent).
- $H_1$: The accused person is guilty (requires evidence beyond a reasonable doubt).

A jury can make two possible errors:

- If they falsely convict an innocent person, that is a false positive and they have made a Type I error.
- If they do not convict a guilty person, that is a false negative and they have made a Type II error.

# <a> Statistical Power </a>
---

The <font color="dodgerblue">**power**</font> of a hypothesis test is <font color="dodgerblue">the probability of correctly rejecting the null hypothesis when it is truly false</font>. If you repeat your experiment (sampling, measuring, and hypothesis test) many times, power is the proportion of times you will reject $H_0$ when it is false. In much simpler terms, it's <font color="dodgerblue">the probability that your study will correctly detect an effect/difference if it exists</font>.

> *"But isn't a hypothesis test giving us us evidence of an effect (or lack of it)?"*
>
> Well, yes, but... We live in a world of uncertainty. We don't know whether a real effect exists in the population or not! If we did know, we wouldn't be testing for it! Practically speaking, we would have to run the same study thousands of times to become more certain. We *may* want to avoid that.
>
>**A study that does not find a statistically significant result does NOT necessarily prove there is no effect**. If a study only had 12 participants, it doesn't prove the treatment doesn't work, it may just mean that the study was not designed well enough to detect the effect.

Maths-wise, power is the opposite of the probability of failing to reject  $H_0$  when it is false, or more succinctly:

$$power = 1 - \beta$$

Imagine we are comparing two distributions using an independent $t$ test. In this case, power is the *probability of detecting a difference between the two distributions if there is one*.  

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/pow_high_low_ex.png?raw=true" width=500/>
</figure>

- If the distributions are well separated (like in a above), our test will have HIGH power because there is a high probability that we will correctly reject $H_0$.
- If the distributions have a lot of overlap (like in b above), our test will have LOW power because the probability of correctly rejecting $H_0$ is lower.

**We may end up rejecting $H_0$ incorrectly when:**
- The samples have high variability
  
<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/samp_var_diff.png?raw=true" width=400/>
</figure>
  
- The effect size is small

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/eff_size_diff.png?raw=true" width=400/>
</figure>

- The sample size is small (smaller samples tend to have more variability)

### <a> Intuition Analogy </a>

Imagine you are fishing in a lake with murky waters, so you can't easily see if fish are there. You cast your net. Sometimes you catch fish, sometimes you don't. If the fish are really there, how likely is your net to catch one?
- A tiny net has low power.
- A large net has high power.

Your study is the net and the fish are the real effect.
- The size of the net itself plays a role, but how long you cast the net for also matters (sample size).
- Some fish can have different sizes or swimming speeds (sampling variability).
- The fish might be very small and slip past your netting, making them difficult to catch in the first place (small effect size).

# <a> Power Analysis </a>
---

Since sampling variability can't be easily controlled and some studies do care about small effect sizes, <font color="mediumseagreen">**the easiest way to increase the statistical power of a test is to increase the sample size**</font>.

<font color="tomato">**Power is a property of study design, not the data you eventually collect.**</font> We typically want to calculate power during the experiment design phase, before carrying out the study. <font color="dodgerblue">The process of calculating what the minimum $n$ should be for our desired power is called **power analysis**</font>. It helps us answer the question *"how many observations / participants / etc do I need in my experiment to be pretty sure I come to the right conclusions?"*

Just like significance levels have a "default" value of 5%, <font color="orange">**the "default" power level researchers tend to aim for is 80%**</font>. This means, for example, that if a treatment truly has an effect, the study has an 80% chance of detecting it. Some studies may aim for a higher power depending on the potential real-world consequences of their findings.



## <a> The Process </a>

The exact calculations can get quite complex and vary based on your statistical test. There are many resources and online calculators to help do the exact calculations in a large variety of cases that you may encounter in the future. <font color="orange">**For this class, we're going to do a simplified shortcut that works for two-tailed independent t-tests that uses the default values for significance and power**</font>.

### Step 1: Define your hypotheses and choose your test

### Step 2: Choose a significance level $\alpha$ and power

For our purposes, we'll use the defaults:
- $\alpha = 0.05$
- 80% power, which makes $\beta = 0.2$

### Step 3: Calculate the $z$ score for $1-\alpha/2$ and $1-\beta$

For our default parameters, that becomes:
- $z_{1 - \alpha/2} = 1.96$
- $z_{1 - \beta} = 0.84$

### Step 4: Estimate the standard deviation of your population ($\sigma$)

This is the tricky step. It's likely impossible to know this since we'd have to measure every single member of the population, so we can estimate it by:
- Using the sample standard deviation from a previous study of a similar population
- Use the sample standard deviation from a pilot study or preliminary data
- Use historical data, if it exists

### Step 5: Choose the smallest significant difference/effect we want to detect ($\delta$)

### Step 6: Calculate the minimum sample size $n$ you need per group

Officially, the equation for this is $\color{dodgerblue}{ \large n = \frac{2 (z_{1-\alpha/2} + z_{1-\beta})^2 \sigma^2}{\delta^2} }$

But, if we use the default values $\alpha = 0.05$ and $\beta = 0.2$ for a two-tailed independent t-test, the entire equation simplifies to:

$\color{dodgerblue}{\large n \approx \frac{16\sigma^2}{\delta^2}}$ in each group.

This simplified equation tells us one important thing: <font color="mediumseagreen">**the smaller the effect you wish to detect is, the larger the $n$ you'll need**</font>.

> <font color="tomato">*It's important to note that this $n$ is an approximation regardless of how it's calculated. You are not guaranteed to detect an effect if it exists, even with 99% power. All you can do is increase your chances!*</font>

# <a name="13intro">Comparing Categorical Data</a>
---

Last week, we explored how we can use t-tests to compare two numerical distributions. <font color="orange">**What if we want to compare two categorical distributions?**</font> There are several tests for categorical data, so we will cover the most common one. It also happens to be a <font color="dodgerblue">**non-parametric test**</font>, meaning it <font color="dodgerblue">does NOT assume that the data comes from a normally distributed population</font>.

## <a> The $\chi^2$ Test of Independence </a>
---

<font color="mediumseagreen">**We use a chi-square ($\chi^2$) test of independence to determine whether two categorical variables are associated in some way**.</font>

Some example questions we can answer with this test:
- Is there a relationship between whether a student has a part time job and their letter grade average?
- Does length of an emergency room stay depend on the severity of a patient's case?



### <a> The basic idea </a>

This test determines whether two categorical variables are related or independent by comparing the *observed* counts in each category (the data we collected) to the counts that we should *expect* if there was no relationship between the variables at all. Under this paradigm, small differences are likely to happen, but large ones may indicate a potential relationship. <font color="tomato">**The $\chi^2$ test of independence can tell you whether a relationship exists, but it CANNOT tell you what *direction* the relationship is in.**</font>

> Note: we may not care whether the data is normally distributed or not, but we do assume that each observation is independent and that the data are counts rather than percentages or ratios.

### <a> What we know </a>

- The observed counts collected in each categorical variable
  - For example, it would look like this if our variables were "treatment group" and "recovery status":
  |  | Recovered | Not Recovered |
  |---|---|---|
  | Treatment A | 32 | 18 |
  | Treatment B | 20 | 30 |

We can then calculate:

- The expected counts $= \Large\frac{(row \,\,total) \cdot (column \,\,total)}{grand \,\,total}$ for each cell in our table.
- The degrees of freedom for our data <font color="tomato">$df = (number\, of \, rows - 1) \cdot (number\, of \, columns - 1)$</font>

### <a> Hypotheses </a>

- **Null hypothesis $H_0$**: the categorical variables are independent, or the variables are NOT associated.
- **Alternative hypothesis $H_1$**: the categorical variables are NOT independent, or the variables ARE associated.

### <a> Test Statistic </a>

For each cell in our table, we calculate $\frac{(observed - expected)^2}{expected}$, then add them all up to get the collective $\chi^2$ statistic:

$$\color{dodgerblue}{\chi^2 = \sum\frac{(observed - expected)^2}{expected}}$$

This $\chi^2$ statistic is the ratio of squared differences with respected to the expected counts. <font color="dodgerblue">**It tells you how different the observed counts are from what you would expect to see if the variables are not associated**</font>.
- It should always be positive.
- A large $\chi^2$ corresponds to a larger difference between observed and expected.
- Unlike the $z$ and $t$ tests whose test statistics are measured in "standard distributions away from the mean", the $\chi^2$ test statistic is unitless.

### <a> The $\chi^2$ Distribution </a>

This is a **right-skewed** distribution specifically used for this test, so it quantifies the likelihood of seeing magnitudes of squared differences. These don't have negative values and it only cares about how large the difference is, so unlike the tests we've done so far, this is a <font color="dodgerblue">**right-tailed test**</font>.

Similar to the $t$ distribution, the exact shape of the $\chi^2$ distribution depends on the degrees of freedom.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/chi_sq_dist.jpg?raw=true" width=400/>
</figure>

You can convert your $\chi^2$ statistic to a $p$-value:
1. Find the probability of seeing a difference between observed and expected counts **at least as extreme as yours** by using the $\chi^2$ distribution table or the R function call <font color="tomato">`pchisq(chi_sq, df, lower.tail = FALSE)`</font>.
2. Unlike previous tests, **you do NOT need to multiply the probability by two** since there is only one tail to account for.

### <a> Critical Value </a>

To find the threshold our test statistic must cross to reject $H_0$, we can:
- Use a $\chi^2$ distribution table for the correct $\alpha$ significance level and degrees of freedom, $t_{\alpha, \, df}$
- Use the <font color="tomato">`qchisq(<alpha>, <df>, lower.tail=FALSE)`</font> R command. **This will give you the upper boundary**. Since the $\chi^2$ distribution is skewed right, the upper and lower boundaries are different, but we only really need the upper boundary.

### <a>Interpreting Results</a>

- A <font size=2>small</font> $\chi^2$ statistic (meaning small differences) will correspond to a large $p$-value, indicating that the differences between observed and expected counts can be attributed to random chance.

- A <font size=5>large</font> $\chi^2$ statistic (meaning large differences) will correspond to a small $p$-value, indicating that the differences between observed and expected counts are unlikely under the null hypothesis.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/chi_sq_test_summ.png?raw=true" width=400/>
</figure>

Comparing your p-value to your $\alpha$ is also very similar to previous tests:

- If your <font color="tomato">**p-value $\leq \alpha$**</font>, it falls within the <font color="tomato">rejection region</font> and we <font color="tomato">**reject $H_0$**</font>.
    - We conclude that there is a relationship between the two variables.
- If your <font color="turquoise">**p-value $> \alpha$**</font>, it falls within the <font color="turquoise">blue region</font> and we <font color="turquoise">**fail to reject $H_0$**</font>.
    - We cannot conclude anything about the relationship between the two variables.
    - Failing to reject $H_0$ does not prove there is no relationship, it just means you do not have strong enough evidence to conclude that there is.

Just like before, remember to write your interpretation within the proper context of the problem and to include the $p$-value.

# <a> $\chi^2$ Test of Independence Example</a>

----

A public health researcher wants to know whether exercise level affects sleep quality in young female adults aged 18-24. She gathers information from 100 unrelated participants on their exercise level (<1 hour/week, 2-4 hours/week, >5 hours/week). She also surveyed their sleep habits and quality, aggregated into a single digit (1: poor, 2: fair, 3: good, 4: excellent).

Following [the steps outlined in the general process](#hypo_test_gen_proc):

### Step 1: Outline what we know and what we don't know

The data collected:

| | <1 hr/wk | 2-4 hr/wk | >5 hr/wk |
|---|---|---|---|
| Sleep: 1 | 9 | 10 | 2 |
| Sleep: 2 | 10 | 7 | 8 |
| Sleep: 3 | 9 | 6 | 9 |
| Sleep: 4 | 3 | 13 | 14 |

### Step 2: Choose the appropriate test
Since we are assessing the relationship between two categorical variables and the observations are independent, we use the chi-square test of independence here.

### Step 3: Choose significance level
Let's use the standard $\alpha$ = 0.05 for this test. Since we are using a $\chi^2$ distribution, we need to find the $\chi^2$-statistic that coresponds to $\alpha$ and the degrees of freedom for our sample.

- Our degrees of freedom $df = (rows - 1) \cdot (columns - 1) = (4-1)\cdot(3-1) = 6$

- Our critical value then is $\chi^2_{\alpha, \, df} = \chi^2_{0.05, \, 6}$
  - We can either use the `qchisq(<alpha>, <df>, lower.tail=FALSE)` function or look it up in a table to find that  $\chi^2_{0.05, \, 6} = 12.59$.

### Step 4: Define your hypotheses
- $H_0$: there is NO association between exercise level and sleep quality (we will assume this is correct and try to disprove it)
- $H_1$: there IS an association between exercise level and sleep quality

### Step 5: Calculate your test statistic

We calculate:
- The expected counts $= \frac{(row \,\,total) \cdot (column \,\,total)}{grand \,\,total}$ for each cell in our table:
  - | EXPECTED | <1 hr/wk | 2-4 hr/wk | >5 hr/wk |
|---|---|---|---|
| Sleep: 1 | 6.51 | 7.56 | 6.93 |
| Sleep: 2 | 7.75 | 9 | 8.25 |
| Sleep: 3 | 7.44 | 8.64 | 7.92 |
| Sleep: 4 | 9.3 | 10.8 | 9.9 |
- The squared differences for each cell $\frac{(observed - expected)^2}{expected}$
  - | EXPECTED | <1 hr/wk | 2-4 hr/wk | >5 hr/wk |
|---|---|---|---|
| Sleep: 1 | 0.95 | 0.79 | 3.5 |
| Sleep: 2 | 0.65 | 0.44 | 0.01 |
| Sleep: 3 | 0.33 | 0.81 | 0.15 |
| Sleep: 4 | 4.27 | 0.45 | 1.69 |
- The collective $\chi^2$ statistic for our data is $\chi^2 = \sum\frac{(observed - expected)^2}{expected} = 14.05 $

This means that the collective measure of how different the observed vs expected counts are is 14.05.

### Step 6: Calculate the $p$-value from the test statistic

We want to know the probability of seeing a difference measure of at least 14.05, so we calculate $P(\chi^2 \geq 14.05)$ using the R function `pchisq()`.

```
pchisq(14.05, df=6, lower.tail=FALSE)
```

We find that $P(\chi^2 \geq 14.05) = 0.0291$, so that is our $p$-value.

### Step 7: Decide whether to reject $H_0$

Two ways to do this:
- Our p-value = 0.0291 is less than $\alpha$ = 0.05, so we reject $H_0$.
- The our test statistic of  14.05 is larger than our critical value of 12.59, so we reject $H_0$.

### Step 8: Interpret the results in the context of our data

There is strong statistical evidence that there is a relationship between exercise level and sleep quality in 18-24 year-old female adults ($p = 0.0291$).

# <a name="hypo_test_gen_proc">Appendix: The General Process of Hypothesis Tests</a>
---

There are MANY different hypothesis tests, and in this class we will only cover a few of them. Still, most hypothesis tests follow the same general approach:

1. Outline what you know about your sample(s) and/or population.

2. Choose the appropriate test based on what you know and the question you want to answer.

3. Choose your <font color="dodgerblue">significance level</font>, the cutoff for "strong enough evidence to reject $H_0$", then convert it to a <font color="dodgerblue">critical value</font> for your test's distribution so you know where the threshold for rejection will be. The default is $\alpha$ = 0.05.

4. *Define the <font color="dodgerblue">null ($H_0$) and alternative ($H_1$) hypotheses</font>*. We will assume that $H_0$ is true and will try to disprove it.
    - $H_0$: "a drug has no effect", "two treatments are equally effective"...
    - $H_1$: "a drug does have an effect", "two treatments have different effect"...

5. *Calculate your <font color="dodgerblue">test statistic</font>*, a single number that measures how far your data is from what $H_0$ predicts. How we calculate this number differs depending on the specific test we're doing, but the idea is the same.

6. *Convert your test statistic to a <font color="dodgerblue">p-value</font>*, the probability of our data being at least this extreme if $H_0$ is true.
    -   The p-value is the area under both tails beyond the test statistic.
    -   The p-value is a probability, so we have $0 < \mbox{p-value} < 1$.
    -   The smaller the p-value, the stronger the evidence to reject $H_0$.

7. *Decide whether to reject $H_0$* or not by either comparing the p-value (our evidence of extremism) to the significance level (the pre-determined area of extremism), or comparing your test statistic (how far your data is from expected) to the critical value (the threshold for extremism).

    - If your <font color="tomato">**p-value $\leq \alpha$**</font>, it falls within the <font color="tomato">rejection region</font> and we <font color="tomato">**reject $H_0$**</font>.
        - This is also the case if <font color="tomato">**abs(test statistic) $\geq$ abs(critical value)**</font> - it means your data has passed the threshold for extremism and therefore have enough evidence to reject $H_0$.
    - If your <font color="turquoise">**p-value $> \alpha$**</font>, it falls within the <font color="turquoise">non-rejection region</font> and we <font color="turquoise">**fail to reject $H_0$**</font>.
        - This is also the case if <font color="turquoise">**abs(test statistic) < abs(critical value)**</font> - it means your data has not passed the threshold for extremism and therefore do not have enough evidence to reject $H_0$.
