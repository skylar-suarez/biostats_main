# <a name="01inference">Introduction to Statistical Inference</a>
---

In biomedical science, we're always trying to answer the "big" questions: what can we learn about different cells? How can we use new information to treat disease?

To really answer these questions, we would need to study every cell, every patient... Unfortunately, doing that is far too costly or even impossible!

- A <font color="dodgerblue">**population**</font> includes *ALL* individuals or objects of interest.
- A <font color="dodgerblue">**sample**</font> is a *subset* of the population.

Some examples:

| Population Example                    | Sample Example                           |
| ------------------------------------- | ---------------------------------------- |
| All patients with Alzheimer’s disease | 50,000 patients with Alzheimer’s disease across the country |
| All liver cells in a patient          | 4,000 liver cells from a patient         |
| All apples from an orchard            | 40 apples from an orchard                |

<br>

So... what do we do if we can't study every single patient or cell?

<br>

# <a>Data Types</a>
---

There are two major data types: quantitative and qualitative.

- <font color="dodgerblue">**Quantitative**</font> variables can be *quantified*, meaning they are ***numerical***. There are two subcategories here:
  - <font color="dodgerblue">**Discrete**</font> variables can only have distinct, countable values. They are typically expressed with integers only. They do NOT include in-between values.
    - Examples: number of cigarettes smoked per day, gene mutation counts.
  - <font color="dodgerblue">**Continuous**</font> variables can have ANY value within a range, including in-between values. They are typically expressed with decimals (`float` in code).
    - Examples: blood pressure readings, height, age.
- <font color="dodgerblue">**Qualitative**</font> variables are ***categorical***, NOT numerical. They may be assigned a "representative" number, but are not truly quantifiable in the same way quantitative data is. There are also two subcategories here:
  - For <font color="dodgerblue">**ordinal**</font> variables, ranking or order matters.
    - Examples: pain severity, cancer stages.
  - For <font color="dodgerblue">**nominal**</font> variables, ranking or order does NOT matter.
    - Examples: sex assigned at birth, blood type, month of the year.

## <a> Descriptive / Summary Statistics for Quantitative Variables </a>
---

- <font color="dodgerblue">**Range**</font>: the the largest and smallest values.
  - The `range(<vector>)` function returns the min and max values in a vector.

- <font color="dodgerblue">**Span**</font>: the distance covered between the smallest and largest values (max - min).
  - Calculate this using `diff(range(<vector>))`.
- <font color="dodgerblue">**Median**</font>: the middle value when the values are arranged in order.
  - Calculate this using `median(<vector>)`.
- <font color="dodgerblue">**Mean**</font>: the arithmetic average of the values; the sum of all values divided by the number of values.
  - Calculate this using `mean(<vector>)`.
- <font color="dodgerblue">**Variance**</font>: how spread out the values are around the mean.
  - Calculate this using `var(<vector>)`.
- <font color="dodgerblue">**Standard Deviation**</font>: square root of the variance.
  - Calculate this using `sd(<vector>)`.

If you're curious about the difference between variance and standard deviation, see [Appendix A](#1_appA).

### Example

Imagine we just measured the resting heart rate in 21 healthy volunteers:

>58, 60, 61, 62, 64, 65, 66, 67, 68, 69, 70, 71, 72, 72, 74, 75, 76, 78, 80, 82, 85

| Statistic                       | Value                | Interpretation                                       |
| ------------------------------- | -------------------- | ---------------------------------------------------- |
| **Minimum**                     | 58 bpm               | Lowest observed heart rate                           |
| **Maximum**                     | 85 bpm               | Highest observed heart rate                          |
| **Range**                       | 85 − 58 = **27 bpm** | Difference between highest and lowest values         |
| **Span**                        | **58–85 bpm**        | Interval covered by the data                         |
| **Median**                      | **70 bpm**           | The middle (11th) observation                        |
| **Mean**                        | **70.2 bpm**         | Average heart rate                                   |
| **Variance** (sample)           | **≈ 58.1 bpm²**      | Variability around the mean                          |
| **Standard Deviation** (sample) | **≈ 7.6 bpm**        | Typical distance from the mean                       |


### <a> Percentiles & Quartiles</a>

A <font color="dodgerblue">**percentile**</font> is a value below which a percentage of observations fall. It splits the data into sections (e.g., the 90th percentile is the value that cuts of the bottom 90% of the data values from the top 10% of data values).

- The 25th percentile is the <font color="dodgerblue">**1st quartile (Q1)**</font> because it represents the first quarter of the data.
- The 50th percentile is the <font color="dodgerblue">**2nd quartile (Q2)**</font> is the **median**, it splits the data in half.
- The 75th percentile is the <font color="dodgerblue">**3rd quartile (Q3)**</font>.

*The "middle" of the data falls within the 25th and 75th percentiles*. The range these values cover is called the <font color="dodgerblue">**interquartile range (IQR)**</font>.

You can calculate a given percentile using `quantile(<vector>, <percentile>)` where the percentile is in decimal form.

You can calculate the IQR using `IQR(<vector>)`.

From our heart rate example above,

>58, 60, 61, 62, 64, 65, 66, 67, 68, 69, 70, 71, 72, 72, 74, 75, 76, 78, 80, 82, 85

| Percentile                      | Value                | Interpretation                                       |
| ------------------------------- | -------------------- | ---------------------------------------------------- |
| **25th**        | **64.5 bpm**           | ~25% of observations are at or below this  |
| **50th**        | ** 70 bpm**          | median, ~50% of observations are at or below this  |
| **75th**        | **75.5 bpm**           | ~75% of observations are at or below this  |
| **90th**             | **82 bpm**           | ~90% of observations are at or below this  |

The IQR of the heart rates is Q3 - Q1, which is $75.5 - 64.5 = 11$ bpm.

## <a> Frequency & Proportion Tables for Categorical Variables </a>
---

When you have categorical data, it's usually more useful to think about it in terms of counts and frequency - how often we see values rather than averages or quartiles.

### <a> Frequency Tables </a>

These show <font color="dodgerblue">*how many observations fall into each category*</font>. They can help compare categories to each other.

Suppose we survey 100 students about their favourite school subject:

| Favourite Subject | Count |
| ---------------- | ----- |
| Math             | 40    |
| Science          | 35    |
| English          | 25    |

From this, we can easily notice that English is the least favourite subject.

In R, you can make a table for a single variable by using the `table(<vector>)` function.

### <a> Proportional Tables </a>

These show the <font color="dodgerblue">*proportion of observations that fall into each category relative to the grand total*</font> instead of raw counts. These help understand the distribution of categories, see patterns and relationships between variables, and estimate probabilities.

Using the same survey of 100 students:

| Favourite Subject | Proportion |
| ---------------- | ---------- |
| Math             | 0.40       |
| Science          | 0.35       |
| English          | 0.25       |

From this, we can infer that if we were to survey a student, there's a 40% chance their favourite subject is math.

In R, you can make a proportional table for a single variable by using the `prop.table(<table>)` function. Note that this function takes a table as input!

> <font color="orange"> Count tables tell you how many, proportional tables tell you how much of the whole. </font>

## <a> Two-Way / Contingency Tables </a>
---

These display the <font color="dodgerblue">*frequencies of observations classified according to two categorical variables at the same time*</font>. The rows represent one variable, the columns represent another, and each cell shows the number of observations that fall into both categories. They help compare groups across categories and investigate possible relationships between variables.

Suppose we survey 100 students again and ask them about whether they play a sport and whether they have a part time job:

|                    | Part-Time Job | No Part-Time Job | Total |
| ------------------ | ------------- | ---------------- | ----- |
| Play a Sport       | 30            | 20               | 50    |
| Don't Play a Sport | 15            | 35               | 50    |
| Total              | 45            | 55               | 100   |

From this table, we learn that 20 students play a sport AND don't have a part time job.

In R, you can make a contingency table by using the `table(<first_vector>,<second_vector>)` function, <font color="tomato">*where the first vector will be represented in the rows and the second vector will be represented by the columns.*</font>

## <a> Proportional Contingency Tables </a>

Just like with a single variable, we can make proportional tables for a two-way table <font color="dodgerblue">relative to the grand total</font>.

Using the same survey of 100 students:

|                    | Part-Time Job | No Part-Time Job | Total |
| ------------------ | ------------- | ---------------- | ----- |
| Play a Sport       | 0.3           | 0.2              | 0.5   |
| Don't Play a Sport | 0.15          | 0.35             | 0.5   |
| **Total**              | 0.45          | 0.55             | 1     |

From this table, we can infer that if we were to survey a student, there's a 15% of them saying they don't play a sport AND have a part-time job.

In R, you can make a proportional contingency table using the `prop.table(<contingency_table>)` function, which works similar to the single-variable call from before.

## <a> Relative Proportional Tables </a>

When we have two variables, things can get more interesting! We can calculate frequencies relative to one variable or the other. This can help us estimate <font color="dodgerblue">**conditional probabilities**</font>. We'll dive further into those later in the class, but for now, consider them as *"if a condition is true, what happens with the other variable?"*.

In our table of 100 students, we had their part-time job status as the columns and their sport status as the rows.

If we calculate the proportional contingency table relative to their sport status, we are calculating the proportions <font color="dodgerblue">**relative to the total in each ROW**</font>. *The first row, which represents all the students that DO play a sport, is then subdivided according to the other variable: whether they have a part time job or not.*

|                    | Part-Time Job | No Part-Time Job | Total |
| ------------------ | ------------- | ---------------- | ----- |
| Play a Sport       | 0.60          | 0.40             | 1.00  |
| Don't Play a Sport | 0.30          | 0.70             | 1.00  |

From this table, we can see that, of all students who do NOT play a sport, 70% of them do NOT have a part-time job. <font color="orange"><u>Pay close attention:</u> in this case, the condition that is true is "student does NOT play a sport", and we're looking at what is happening in the other variable, part-time job status.</font>

In R, you can make a proportional table relative to the row total using the `prop.table(<contingency_table>, 1)` function.

We can also calcualte the proportional contingency table <font color="dodgerblue">**relative to the total in each COLUMN**</font>, which is their sport status. Similarly, *the first column, which represents all the students that DO play a sport, is then subdivided according to the other variable: whether they have a part-time job or not.*

|                    | Part-Time Job | No Part-Time Job |
| ------------------ | ------------- | ---------------- |
| Play a Sport       | 0.667         | 0.364            |
| Don't Play a Sport | 0.333         | 0.636            |
| **Total**              | 1.000         | 1.000            |

From this table, we can see that, of all students who DO have a part-time job, 33.3% of them do NOT play a sport. <font color="orange">In this case, the condition that is true is "student DOES have a part time job", and we're looking at what's happening in the other variable, sport status.</font>

In R, you can make a proportional table relative to the column total using the `prop.table(<contingency_table>, 2)` function.

## <a> Boxplots </a>
---

There are many ways to visually summarize numerical data. Boxplots are especially useful because they give an easy visual of spread, center, and unusual values all in one.

To create a boxplot:

-   Find the values of $Q_1$, median, and $Q_3$.
-   Draw a box with edges at $Q_1$ and $Q_3$ and line inside the box for the median. The box here essentially denotes the boundaries of the interquantile range (IQR).
-   Identify the upper and lower fence to classify outliers:
    -   Upper fence $=Q_3 + 1.5(\mbox{IQR})$.
    -   Lower fence $=Q_1 - 1.5(\mbox{IQR})$.
-   Extend a line (whisker) from the lower edge of box to the smallest observation greater than the lower fence.
-   Extend a line (whisker) from the upper edge of the box to the largest
    value that is less than the upper fence.
-   The observations that are less than the lower fence or greater than
    the upper fence are considered
     <font color="dodgerblue">**outliers** </font> and are marked with individual points.

Remember our heart rate data from 21 healthy volunteers? Here's what that boxplot looks like:


```r
options(repr.plot.width = 6, repr.plot.height = 3) # adjust plot window size

healthy_hr <- c(58, 60, 61, 62, 64, 65, 66, 67, 68, 69, 70, 71, 72, 72, 74, 75, 76, 78, 80, 82, 85)
boxplot(healthy_hr,
        horizontal=TRUE,
        main="heart rates of 21 healthy volunteers",
        xlab="heart rate (bpm)")
```

## <a> Skew & Spread </a>
---

The shape of a distribution can tell us a lot about how the values are spread out. The easiest to spot visually is <font color="dodgerblue">**skew**</font>, which describes the direction of the tail of the data. The tail of the data indicates the direction of outliers (if any).

We consider the <font color="magenta">median</font> and the <font color="orange">mean</font> to be the measures of centrality of continuous data. <font color="dodgerblue">*Since the mean is the average of all values, it gets "pulled" by outliers.*</font>

<figure>
<img src="https://github.com/skylar-suarez/Biostats_main/blob/main/Images/skews.png?raw=true" width=600/>
</figure>

- If your data has a <font color="dodgerblue">**positive skew**</font>, the outliers are on the right (greater) side, so your mean will get pulled in that direction, making the <font color="magenta">mean</font> **>** <font color="orange">median</font>.
- If you data is <font color="dodgerblue">**symmetrical**</font>, most of the data is in the center, so the mean and median will be very similar.
- If your data has a <font color="dodgerblue">**negative skew**</font>, the outliers are on the left (lesser) side, so your mean will get pulled in that direction, making the <font color="magenta">mean</font> **<** <font color="orange">median</font>.

Another attribute we can easily spot by eye is <font color="dodgerblue">**variability**</font> - how different the values are with respect to each other or their average. <font color="dodgerblue">*Generally speaking, a narrow and tall distribution has less variability than a flat and wide distribution*</font>.

Earlier in this notebook, we said that the standard deviation is the measure of how spread out values are on average. A smaller standard deviation means the values are less far away from the mean, and therefore have less variability - in other words, they are less "different" from the mean.

# <a name="1_appA"><font color="orange"> Appendix A: Why both variance and standard deviation?</font></a>
---

Variance and standard deviation may seem like they fulfill the same purpose at first glance. Even though they both describe variability, they serve different purposes.

<font color="dodgerblue">**Variance**</font> measures the average squared distance from the mean.
- Squaring makes sure that the positive and negative distances don't cancel each other out.
- Variance has squared units ($bpm^2, kg^2$).
- It's mathematically convenient for most statistical calculations and methods.

<font color="dodgerblue">**Standard deviation**</font> is the square root of the variance.
- It has the same units as the original data ($bpm, kg$).
- It is easier to interpret because it represents a "typical" distance from the mean.

From the heart rate example:
- Variance is ~ $58.1 \,bpm^2$
  - This may be communicated as *"the participants' heart rates had $58.1 bpm^2$ of variability."*
- Standard deviation is ~ $7.6\, bpm$
  - This may be communicated as *"the participants' heart rates were on average 7.6 bpm away from the mean."*

The latter is easier to communicate than the former.

> <font color="orange">*Variance is the engine under the hood, but standard deviation is the dashboard display.*</font>
