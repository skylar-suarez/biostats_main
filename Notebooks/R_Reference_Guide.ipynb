Storing something as a variable will not automatically print it out. You will have to call the variable on its own line to print it out, like below:

# <a name="_">Data Types</a>
---

R has 6 basic [data
types](https://cran.r-project.org/doc/manuals/r-release/R-lang.html#Basic-types):

1.  <font color="dodgerblue">**character**</font>: collections of characters. E.g., `"a"`, `"hello world!"`
2.  <font color="dodgerblue">**double**</font>: decimal numbers. e.g., `1.2`, `1.0`
3.  <font color="dodgerblue">**integer**</font>: whole numbers. In R, you must add `L` to the end of a number to specify it as an integer. E.g., `1L` is an integer but `1` is a double.
4.  <font color="dodgerblue">**logical**</font>: Boolean values, `TRUE` and `FALSE`
5.  <font color="dodgerblue">**complex**</font>: complex numbers. E.g., `1+3i`
6.  <font color="dodgerblue">**raw**</font>: a type to hold raw bytes.

See [Fundamentals of Working with
Data](https://githubtocolab.com/CU-Denver-MathStats-OER/Statistical-Theory/blob/main/Appendix/Intro-to-Vectors-Dataframes.ipynb)
for more information.
## `Factor` data type

The `factor` data type is categorical and has a fixed set of possible values called *levels*. Instead of treating data as plain numbers, `factor` tells R that the values represent categories instead of numerical values. This is important for modeling and tables!

To create a new categorical variable, you can call the `factor()` function with a collection of levels:





```r
seasons <- c("spring", "summer", "fall", "winter")
seasons <- factor(seasons)

# or simply
seasons <- factor(c("spring", "summer", "fall", "winter"))
seasons
```

You can also convert a numerical vector into a factor, and even give it custom string levels:

```r
seasons_num <- c(1, 2, 3, 4)

# convert to factor (numerical values as the levels)
seasons_fac <- factor(seasons_num)
seasons_fac

# convert to factor with custom levels
seasons_str <- factor(seasons_num, labels=c("spring", "summer", "fall", "winter"))
seasons_str
```

```r
x <- 10
y <- 5

x != y # should be TRUE
x <= y # should be FALSE
(x > y) & (x <= y) # should be FALSE (second one is false)
(x > y) | (x <= y) # should be TRUE (first one is true)
!(x != y) # should be FALSE because the inside is true
```

# <a name="_">Tables for Categorical Variables</a>
---



| Description          | Function              | Description            |
| -------------------- | --------------------- | ---------------------- |
| Frequency table of dataframe      | `table(df)`            | Counts the instances of each value in the dataframe       |
| Frequency table of vector     | `table(v)`            | Counts the instances of each value in the vector      |
| Two-way frequency table | `table(rowvec, colvec)` | Counts the instances of each value in the two vectors |
| Grand Proportion Table | `prop.table(table)` | The input *must be a table* |
| Relative Frequency Table (row) | `prop.table(table, 1)` | Gives the proportions relative to *row* totals |
| Relative Frequency Table (columns) | `prop.table(table, 2)` | Gives the proportions relative to *column* totals |

Pay special attention when creating relative tables - they will be based on the rows and columns from the table you give it!






```r
# example: apples produced by a tree over many seasons
apples <- c(10, 20, 40, 0, 10, 30, 30, 10, 20, 30, 40)
season <- factor(c("spring", "summer", "fall", "winter", "spring", "summer", "fall", "winter", "spring", "fall", "fall"))
my_df <- data.frame(season=season, apples=apples)

print("frequency table for the whole dataframe")
table(my_df)

print("frequency table for the apples column")
table(my_df$apples)

print("frequency table for apples (rows) and season (columns)")
my_tbl <- table(my_df$apples, my_df$season)
my_tbl

print("overall proportion table")
prop.table(my_tbl)

print("relative frequency table relative to ROW totals")
prop.table(my_tbl, 1)

print("relative frequency table relative to COLUMN totals")
prop.table(my_tbl, 2)
```

# <a> Distribution Functions </a>
---

## Binomial

-   `dbinom(x, n, p)` calculates the probability of <font color="dodgerblue">**exactly** $x$ successes</font> out $n$ trials, $\color{dodgerblue}{p(x)=P(X=x)}$.
-   `pbinom(x, n, p)` calculates the probability of <font color="tomato">**at most** $x$ successes</font> out $n$ trials, $\color{tomato}{F(x)=P(X \leq x)}$.
-   `rbinom(m, n, p)` <font color="dodgerblue">randomly sample</font> $m$ values from $X \sim \mbox{Binom}(n, p)$ (with replacement).
-   `qbinom(q, n, p)` compute the <font color="dodgerblue">q<sup>th</sup> quantile</font>.



## Poisson

-   Use `dpois(x, lambda)` to calculate the probability of <font color="dodgerblue">**exactly** $x$ occurrences $p(x) = P(X = x)$</font>.
-   Use `ppois(x, lambda)` to calculate the probability of <font color="tomato">**at most** $x$ occurences $F(x) = P(X \leq x)$</font>.
-   Use `rpois(m, lambda)` to <font color="dodgerblue">randomly sample </font> $m$ values from $X \sim \mbox{Pois}(\lambda)$.
-   Use `qpois(q, lambda)` to find the <font color="dodgerblue">q<sup>th</sup> quantile</font>.

## Normal / Gaussian & Standard Normal Distributions

To work from the standard normal distribution, use `mu = 0` and `sigma = 1`.

-   Use `dnorm(x, mu, sigma)` to calculate the **height** of the PDF at $x$ (not the probability!).
-   Use `pnorm(x, mu, sigma)` to calculate the <font color="dodgerblue">probability of **up to and including $x$**, $P(X \leq x)$</font>
    - This is the area to the *left* of $x$.
-   Use `pnorm(x, mu,  sigma, lower.tail=FALSE)` to calculate the <font color="dodgerblue"> probability of **above $x$**, $P(X>x)$</font>
    - This is the area to the *right* of $x$.
- Use `rnorm(m, mu, sigma)` to <font color="dodgerblue">randomly sample</font> $m$ values from ${X \sim N( \mu ,\sigma)}$
- Use `qnorm(q, mu, sigma, lower.tail=TRUE)` to find the <font color="dodgerblue">q<sup>th</sup> quantile</font>.

## Students' $t$ Distribution

-   Use `dnt(x, df)` to calculate the **height** of the PDF at $x$ (not the probability!).
-   Use `pt(x, df)` to calculate the <font color="dodgerblue">probability of **up to and including $x$**, $P(X \leq x)$</font>
    - This is the area to the *left* of $x$.
-   Use `pt(x, df, lower.tail=FALSE)` to calculate the <font color="dodgerblue"> probability of **above $x$**, $P(X>x)$</font>
    - This is the area to the *right* of $x$.
- Use `rt(m, df)` to <font color="dodgerblue">randomly sample</font> $m$ values from ${X \sim N( \mu ,\sigma)}$
- Use `qt(q, df, lower.tail=TRUE)` to find the <font color="dodgerblue">q<sup>th</sup> quantile</font>.

# <a> Hypothesis Tests </a>
---

These functions return an object with useful fields like `$statistic`, `$parameter`, `$p.value`, and `$alternative`. To access them, store the returned object in a variable first.

## <a> Parametric Tests </a>

- Two-tailed PAIRED $t$ test: <font color="mediumseagreen">`t.test(<first_vector>, <second_vector>, paired=TRUE, alternative="two.sided")`</font>
- Two-tailed INDEPENDENT $t$ test of EQUAL variance: <font color="mediumseagreen">`t.test(<first_vector>, <second_vector>, var.equal=TRUE, alternative="two.sided")`</font>
- $\chi^2$ test of independence:
  1. Make a contingency table of your variables (e.g., `table(<first_vector>,<second_vector>)`).
  2. Use the function call <font color="mediumseagreen">`chisq.test(<contingency_table>)`</font>.
- Two-tailed $F$ test of equal variance: <font color="mediumseagreen">`var.test(<first_vector>, <second_vector>, alternative="two.sided")`</font>
- ANOVA:
  - If your data is is contained in a numeric vector for each group, you need to combine them into a factor object (example below), then use the function call <font color="mediumseagreen">`aov.test(<observations> ~ <groupings>)`</font>.
  - If your data is in a dataframe that is separated by group, use the function call <font color="mediumseagreen">`aov.test(<observations> ~ <groupings>, data=<dataframe>)`</font>.

## <a> Normality Tests </a>

- Shapiro-Wilk test: <font color="mediumseagreen">`shapiro.test(<variable>)`</font>

## <a> Non-Parametric Tests </a>

- Two-tailed Wilcoxon Rank Sum test: <font color="mediumseagreen">`wilcox.test(<first_vector>, <second_vector>, alternative="two.sided")`</font>
- Welch's independent $t$ test: <font color="mediumseagreen">`t.test(<first_vector>, <second_vector>, var.equal=FALSE, alternative="two.sided")`</font>

# <a> Basic Plotting Functions </a>
---

The `plot()` function is the most broadly used function for plotting different data types.
-   If we input one categorical variable that is stored as `factor`, we get a bar chart.
-   If we input one quantitative variable, it will make a scatter plot.

```r
par(mfrow = c(1, 2))  # create a row of 2 plots

plot(my_df$season,  # categorical data, so it will make a bar plot
     main = "Frequency of Seasons",  # main title
     xlab = "Season",  # horizontal axis label (will cut off name if it doesn't fit)
     ylab = "Frequency",  # vertical axis label
     col = "steelblue")  # fill color of bars

plot(my_df$apples,  # quantitative data, so it will make a scatter plot
     main = "Apple Production",  # main title
     xlab = "Index (Row of Observation)",  # horizontal axis label
     ylab = "# of apples produced")  # vertical axis label
```

## Basic Bar and Pie Charts

These both take tables as input!

```r
par(mfrow = c(1, 2))

apple_table <- table(my_df$apple)  # create table of apple production counts
apple_table

barplot(apple_table,  # apple counts
        main = "Frequency of Apple Production",  # main title
        xlab = "Apple Production",  # horizontal axis label
        ylab = "Frequency",  # vertical axis label
        col = "seagreen")  # fill color of bars

pie(apple_table,  # input table of category counts
    main = "Frequency of Apple Production")  # main title
```

## Grouped Frequency Bar Charts

To create a bar chart that displays one variable in relationship to another:
1. Create a two-way frequency table for the variables you want. It will take the **columns** of your table as its x-axis.
2. Use that as the input to the `barplot()` function.

**Note: Pay close attention to how the table was made - what rows and columns represent.**

```r
my_tbl <- table(my_df$apples, my_df$season) # num of apples in rows, seasons in cols
my_tbl

# create side by side bar chart
barplot(my_tbl,  # use counts from the table we made
        beside = TRUE,  # groups side-by-side
        main = "# of apples produced per season",  # main title
        xlab = "Month",  # horizontal axis label
        col = c("gray", "green", "purple", "red", "blue"),  # fill color of bars
        ylab = "Frequency")  # vertical axis label

# add a legend to plot
legend(x="topleft",  # place legend in top left
       legend=rownames(my_tbl),  # get labels from row name in the table
       fill = c("gray", "green", "purple", "red", "blue"))  # use same fill colors
```

## Stacked Relative Frequency Bar Charts

To create a bar chart displaying the relative frequency (or proportion)
of category hurricanes that occurred in each month:

1.  First create a two-way table of relative frequencies. It will take the **columns** of your table as its x-axis.
  -   Pay attention to whether you want the proportions relative to grand, row, or column totals.

2.  Input the table into the `barplot()` function.
  -   The default option `beside = FALSE` stacks the bars.

```r
my_tbl <- table(my_df$season, my_df$apples) # seasons in rows, num of apples in cols

seasons_grand <- prop.table(my_tbl)
seasons_grand

barplot(seasons_grand,  # use proportions from contingency table
        main = "Apple Production by season (overall proportions)",  # main title
        xlab = "# of apples produced",  # horizontal axis label
        col = c("salmon", "lightblue", "orange", "aquamarine"),  # color of bars
        ylab = "Proportion")  # vertical axis label
legend(x="topright",  # place legend in top left
       legend=rownames(seasons_grand),  # get labels
       fill = c("salmon", "lightblue", "orange", "aquamarine"))  # use same colors
```

```r
rel_to_seasons <- prop.table(my_tbl, 1)
rel_to_seasons

barplot(rel_to_seasons,  # use proportions from contingency table
        main = "Apple Production by season (relative to row totals)",  # main title
        xlab = "Season",  # horizontal axis label
        col = c("salmon", "lightblue", "orange", "aquamarine"),  # color of bars
        ylab = "Proportion")  # vertical axis label
legend(x="topright",  # place legend in top left
       legend=rownames(rel_to_seasons),  # get labels
       fill = c("salmon", "lightblue", "orange", "aquamarine"))  # use same colors
```

```r
rel_to_apples <- prop.table(my_tbl, 2)
rel_to_apples

barplot(rel_to_apples,  # use proportions from contingency table
        main = "Apple Production by season (relative to column totals)",  # main title
        xlab = "Season",  # horizontal axis label
        col = c("salmon", "lightblue", "orange", "aquamarine"),  # color of bars
        ylab = "Proportion")  # vertical axis label
legend(x="topright",  # place legend in top left
       legend=rownames(rel_to_apples),  # get labels
       fill = c("salmon", "lightblue", "orange", "aquamarine"))  # use same colors
```

### Grouped Histograms

```r
# Generate sample data
set.seed(42)
x1 <- rnorm(100)
x2 <- rnorm(100, mean = 1.5)
x3 <- rnorm(100, mean = -1)

# 1. Calculate combined limits
h1 <- hist(x1, plot = FALSE)
h2 <- hist(x2, plot = FALSE)
h3 <- hist(x3, plot = FALSE)

min_x <- min(h1$breaks, h2$breaks, h3$breaks)
max_x <- max(h1$breaks, h2$breaks, h3$breaks)
max_y <- max(h1$counts, h2$counts, h3$counts)

# 2. Create the blank rescaled plot
plot(1, type = "n",
     main = "title",
     xlim = c(min_x, max_x),
     ylim = c(0, max_y),
     xlab = "Combined X Values",
     ylab = "Frequency")

# 3. Choose the colors
colors <- c(rgb(1, 0, 0, 0.4),  # Red with transparency
            rgb(0, 1, 0, 0.4),  # Green with transparency
            rgb(0, 0, 1, 0.4)) # Blue with transparency

# 4. Add the histograms
plot(h1, col = colors[1], add = TRUE)
plot(h2, col = colors[2], add = TRUE)
plot(h3, col = colors[3], add = TRUE)

# 5. Add the legend
legend("topright", c("x1", "x2", "x3"), fill=colors)
```

- To add a legend, use the `legend()` function call immediately after plotting.
- To make a plot grid (for side by side plots, for example), use `par(mfrow = c(rows, cols))` before your plotting functions.
  - This does NOT change the default size of the figure! To do that, add `options(repr.plot.width = 10, repr.plot.height = 5)` (with your desired width and height) before the plotting functions.
    - This change persists in subsequent code cells. To reset it to the default, use the same function call with both width and height set to 6.
