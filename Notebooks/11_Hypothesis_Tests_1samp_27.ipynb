# <a name="23intro">Hypothesis Testing</a>

---

Hypothesis tests are, in a way, the heart of statistical data analysis. If you have a hypothesis, you generally have to test it before making any claims that it is correct, and to do that, we need some evidence. Hypothesis tests are a way to do exactly that.

A <font color="dodgerblue">**hypothesis test**</font> is a <font color="dodgerblue">structured way to use data to evaluate claims, balancing evidence against the possibility of random variation</font>. It helps us make decisions or draw conclusions about a population.

Like we discussed last week with confidence intervals, we usually do not have access to an entire population, only a sample of it. Hypothesis tests *help us decide whether the patterns we see in that sample are strong enough to conclude something real is happening* or whether they are due to random chance.

There are many different types of hypothesis tests that each test for different things. In fact, a lot of statistics functions you may have used in Python or other languages are likely doing hypothesis tests under the hood!

There are two major categories of hypothesis tests: **parametric** and **non-parametric**.
- <font color="dodgerblue">**Parametric**</font> tests *assume data follows a specific distribution*, using parameters like mean, standard deviation, etc. These tests are very useful for continuous, normal data.
- <font color="dodgerblue">**Non-parametric**</font> tests *do NOT assume anything about the data distributions*. These tests are more useful for skewed or very small datasets.

### For now, we will focus on parametric tests - meaning, <font color="dodgerblue">**our tests will assume that the data you are testing is approximately normal.**</font>




# <a> Key Components of Hypothesis Tests</a>
---

Before we dive into how to actually conduct a hypothesis test, we should cover the key components to understand what calculations we will need to do.

First, the basics:

- **Information on sample and context**: what group or situation we're dealing with.
    - E.g., "a sample of 30 male patients aged 50-60 years old"
- **Variables**: the measurable elements involved (independent and dependent variables).
    - E.g., "low-sugar diet (independent) may affect blood glucose levels (dependent)"
- **Claim or relationship**: A specific statement about how the variables relate - *the thing we want to test for!*
    - E.g., "switching to a low-sugar diet changes the blood glucose levels of 50-60 year old males"

So how do we test the claim? We need to construct **hypotheses** that we can test directly. A hypothesis test usually involves two hypotheses:
- The <font color="dodgerblue">**Null Hypothesis ($H_0$)**</font> is the *default assumption* - typically, this is "no difference" or "no effect". <font color="mediumseagreen">In hypothesis testing, we assume that $H_0$ is true, then try to disprove it</font>.
    - E.g., "the blood glucose levels of 50-60 year old males on a low-sugar diet is **the same** as their blood glucose levels were prior to switching from a regular diet"
- The <font color="dodgerblue">**Alternative Hypothesis ($H_1$)**</font> is *the claim you are testing*, usually the opposite of $H_0$ - typically, this is "there is a difference" or "there is an effect".
    - E.g., "the blood glucose levels of 50-60 year old males on a low-sugar diet is **NOT the same** as their blood glucose levels were prior to switching from a regular diet"

> <font size="2">You might notice that the null hypothesis that we are testing is different from the claim made at the beginning. This is because it's much easier to assess evidence against a claim than to definitively prove it true. For more information and intuition behind this, see [Appendix A](#11appendixA).</font>

So how do we try to disprove $H_0$? We need a few more things...

- <font color="dodgerblue">**Test statistic**</font>: A single-number summary of your data that measures how far your data is from what you should expect to see.
    - *This is calculated differently depending on the test*. For example, if your data is normal and has a large sample size, you might calculate a **z-score**, while if you have a small dataset, you might calculate a **t-statistic**. We will explore more of these as we move through the course.

The test statistic alone is not enough to interpret whether our data has statistical significance or not. For that, we need two other things: a <font color="dodgerblue">**p-value**</font> and a <font color="dodgerblue">**significance level**</font>. More on these in a bit!

# <a> Components of Hypothesis Tests</a>
---

As we mentioned before, the test statistic does tell us how far our data is from the expected value, but it's not enough to decide whether what effect/difference we're seeing is "real" or due to random chance. Both significance levels and p-values help answer these questions - they tell us whether our findings are *statistically significant*.

### <a>Test Statistic</a>

The <font color="dodgerblue">**test statistic**</font> tells us *how far your observed sample mean is from what we would expect under the null hypothesis*. The calculation is generally fairly simple:

$$\color{dodgerblue}{\,\,\,\, test\,statistic = \frac{difference\,between\,sample\,and\,expected\,means}{standard\,error\,of\,the\,sample\,mean}}$$

Two factors influence the test statistic: how far your sample mean is from what you expected to see, and how precisely you estimate the mean.

If you remember from last week, the standard error tells us how much a sample mean varies from sample to sample when we repeatedly take samples from the same population.
- A larger standard error will yield a smaller test statistic.
- A larger sample size will shrink the standard error and therefore yield a larger test statistic.

### <a>$p$-value<a>

A <font color="dodgerblue">**$p$-value**</font> answers the question: *"if $H_0$ were true, how rare is the data we observed?"* by telling you the **probability** of getting the data you observed. You can think of it as a measure of "extremeness", or how strong the evidence against $H_0$ is. <font color="mediumseagreen">**When conducting a hypothesis test, this will be the value you report alongside your conclusions.**</font>

For example, if your $p$-value is 0.15, it means that your sample is in the 15% most extreme samples we would expect to see if $H_0$ is true.

- A ***small*** p-value → data is **unlikely** under the assumption that $H_0$ ("there is no difference") is true.
- A ***large*** p-value → data is likely due to random chance.


<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/p_val_int.png?raw=true" width=500/>
</figure>

As an example, imagine flipping a fair coin 100 times:
- If you get heads 55 times, that's **not rare** since the probability of that is pretty high - in fact, that's pretty close to what you would expect to see. **In this case, the p-value is large.**
- If you get heads 97 times, that's **very rare** since the probability of that is very low. **In this case, the p-value is small**.

These concepts are visually explained in the picture below.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/hypotest_pval.png?raw=true" width=500/>
</figure>

We get this p-value by <font color="dodgerblue">converting our test statistic into the probability of seeing values at least as extreme as that within the probability distribution for our test</font> and multiplying it by 2 (to account for the two tails of the distribution).

- Using our "blood glucose changes with a low-sugar diet" example, imagine you collect your data and find out that your test statistic is -1.37. This means that your data is -1.37 standard deviations away from what you expected to see. To translate this into a probability, we look it up and find that the probability of seeing values that far out from the center is $\approx$0.09 x 2 $\approx$ 0.18.

But it's not just enough to have the probability of seeing our data - we have to compare it against something to decide whether that probability is "significant" enough to make a judgement on it. Enter...

### <a>Significance Level</a>

A <font color="dodgerblue">**significance level $\alpha$**</font> in a hypothesis test is similar to a confidence interval's $\alpha$. It is a sort of threshold you choose that determines when to reject $H_0$ - we are essentially choosing how extreme our data would have to be for us to consider it enough evidence to reject $H_0$. Another way to think of it is  how small the $p$-value would have to be to say that the evidence against $H_0$ is enough to reject it.</font>.

But how do we know whether our data is within that area of extremesism? We need to set a numerical threshold - an actual boundary point that determines the when to reject $H_0$. The value within the distribution that marks this cutoff point is called the <font color="dodgerblue">**critical value**</font>.
- For example, if your test involves a normal distribution, this would be the z-score that corresponds to your $\alpha$ value (e.g., for $\alpha$ = 0.05, the critical value would be $z_{\alpha/2}$ = $z_{0.025}$ = 1.96)

These concepts are visually explained in the picture below.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/hypotest_siglev.png?raw=true" width=500/>
</figure>

Just like with confidence intervals, <font color="dodgerblue">*the most widely used $\alpha$ is **0.05***</font> (split into two tails, on either end). This means that our data will have to be in the 5% most extreme values possible for us to consider it enough evidence to reject $H_0$.

<font color="red" size="4">**If you are not explicitly given a significance level for your test, always assume that alpha is 0.05.**</font>

### <a>Two-Tailed Testing</a>

When we split the alpha into the two tails like this, we call it a <font color="dodgerblue">**two-tailed** hypothesis test</font>. **A two-tailed test looks for extreme results in both directions**, so *unusually high or unusually low observations can both count as evidence against $H_0$*.

<font color="mediumseagreen">*Note: The significance level is not something we compute, it's something we choose prior to our analysis. Do NOT first calculate the p-value, and then retroactively choose the significance level to ensure the result is significant!*</font>



# Putting it all together

Since the significance level $\alpha$ gives us the threshold that our observation must cross to be considered "too extreme" (the critical value), we call the area under the curve that represents that probability the <font color="tomato">**rejection region**</font>.

**If our observation's p-value falls within the rejection region, we can reject $H_0$ since the evidence is telling us that our observation is too unlikely to be due to random chance.**

So... Once we collect our data and find our data's test statistic, we can convert it to a p-value and we can see whether it falls within the rejection region or not.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/hypo_test_sum.png?raw=true" width=500/>
</figure>

If it does, we have enough evidence to say that $H_0$ is very unlikely to be true (e.g., "there is a difference!")

> In our [courtroom analogy](#analogy), the complete phrase is actually <font color="orange">*"innocent until proven guilty **beyond reasonable doubt**"*</font>. <font color="dodgerblue">The $p$-value is the evidence, and the significance level is the threshold the evidence must cross to be considered " beyond reasonable doubt"</font>.

## <a>Summarizing the Results</a>

Numerically, the way we can tell whether the p-value is within the rejection region or not is by comparing the p-value to $\alpha$. There are two possible outcomes:

- If your <font color="tomato"> **p-value $\leq \alpha$**</font>, it falls within the <font color="tomato">red region</font> and we <font color="tomato">**reject $H_0$**</font>. This means that our test is *statistically significant*, and we accept the competing claim made in $H_1$.
- If your <font color="turquoise">**p-value $> \alpha$**</font>, it falls within the <font color="turquoise">blue region</font> and we <font color="turquoise">**fail to reject $H_0$**</font>. *This means our test is inconclusive regarding $H_0$ and $H_1$*.

In the end, we want to be sure we communicate the results **clearly**,
in **proper context**, to a more **general audience** that may not have
an advanced background in statistics and mathematics.

When we report our findings, <font color="mediumseagreen">we always want to state how $p$-value compared with the $\alpha$</font>.
- For example, if we used $\alpha$ = 0.05 and our $p$-value fell in the rejection region, we would add something like `(p < 0.05)` at the end of our statement, or even the value itself like `(p = 0.43)`. In scientific publications, you wil often see this written within the interpretation, as well as indicated somewhere in a figure or graph (e.g., an asterisk * for 0.01 $\leq p \leq$ 0.05)

# <a name="hypo_test_gen_proc">The General Process of Hypothesis Tests</a>
---

There are MANY different hypothesis tests, and in this class we will only cover a few of them. Still, most hypothesis tests follow the same general approach:

1. Outline what you know about your sample(s) and/or population.

2. Choose the appropriate test based on what you know and the question you want to answer.

3. Choose your <font color="dodgerblue">significance level</font>, the cutoff for "strong enough evidence to reject $H_0$", then convert it to a <font color="dodgerblue">critical value</font> for your test's distribution so you know where the threshold for rejection will be. The default is $\alpha$ = 0.05.

4. *Define the <font color="dodgerblue">null ($H_0$) and alternative ($H_1$) hypotheses</font>*. We will assume that $H_0$ is true and will try to disprove it.
    - $H_0$: "a drug has no effect", "two treatments are equally effective"...
    - $H_1$: "a drug does have an effect", "two treatments have different effect"...

5. *Calculate your <font color="dodgerblue">test statistic</font>*, a single number that measures how far your data is from what $H_0$ predicts. How we calculate this number differs depending on the specific test we're doing, but the idea is the same.

6. *Convert your test statistic to a <font color="dodgerblue">p-value</font>*, the probability of our data being at least this extreme if $H_0$ is true.
    -   The p-value is the area under both tails beyond the test statistic.
    -   The p-value is a probability, so we have $0 < \mbox{p-value} < 1$.
    -   The smaller the p-value, the stronger the evidence to reject $H_0$.

7. *Decide whether to reject $H_0$* or not by either comparing the p-value (our evidence of extremism) to the significance level (the pre-determined area of extremism), or comparing your test statistic (how far your data is from expected) to the critical value (the threshold for extremism).

    - If your <font color="tomato">**p-value $\leq \alpha$**</font>, it falls within the <font color="tomato">rejection region</font> and we <font color="tomato">**reject $H_0$**</font>.
        - This is also the case if <font color="tomato">**abs(test statistic) $\geq$ abs(critical value)**</font> - it means your data has passed the threshold for extremism and therefore have enough evidence to reject $H_0$.
    - If your <font color="turquoise">**p-value $> \alpha$**</font>, it falls within the <font color="turquoise">non-rejection region</font> and we <font color="turquoise">**fail to reject $H_0$**</font>.
        - This is also the case if <font color="turquoise">**abs(test statistic) < abs(critical value)**</font> - it means your data has not passed the threshold for extremism and therefore do not have enough evidence to reject $H_0$.

<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/hypo_test_interp.png?raw=true" width7500/>

8. Interpret and summarize the results of our test within the context of the original question, including the $p$-value.


> <font color="mediumseagreen">In short, we assume "no effect/difference", check how unlikely our data would be under that assumption, and then decide whether our data is extreme enough to question the "no effect/difference" claim.</font>

# <a> The two-tailed 1-sample $z$-test </a>
---

This is the first (and simplest) type of hypothesis test we'll cover.

### <a>The basic idea</a>

We use a two-tailed 1-sample $z$-test when we want to answer the question:

> <font color="dodgerblue">*"Is my sample mean meaningfully different from a claimed/known population mean value?"*</font>

Some example questions we can answer with this test:
- "Is the cell division rate of my cancerous skin cell culture different from the established average cell division rate of non-cancerous skin cells?"
- "Is the protein content of my shipment of granola bars actually what the company claims it is?"
- "Is the blood glucose level of young adult women on a low-sugar diet lower than the national average?"

### When to use this test

To use a two-tailed 1-sample $z$-test properly, we must:
- Have data from a <font color="tomato">**relatively large sample ($n \geq 200$)**</font> that is <font color="tomato">approximately normal</font>, and
- Know the <font color="tomato">**true/claimed population mean $\mu$**</font> and<font color="tomato"> **standard deviation $\sigma$**</font> that we are comparing our sample to.  

### <a>What we know</a>

- Your sample observations $x$
    - The sample mean $\bar{x}$
    - The sample size $n$
- The population mean value $\mu$
- The population standard deviation $\sigma$

### <a>Hypotheses</a>

The notation here gets a little tricky, so pay close attention... At this point, there are two quantities we care about:
- The true/claimed population mean $\mu$, which we should know to use this test
- The sample population mean $\mu_0$
    - This represents our best estimate of the true population mean, which in our case is our sample mean $\bar{x}$.
    - Why not just write $\bar{x}$ here? Short answer: notation rules.

Knowing that, we write our hypotheses:

- **Null hypothesis $H_0$**: $\mu_0 = \mu$ (meaning estimated our sample mean $\mu_0$ is the same as the population mean value $\mu$)
- **Alternative hypothesis $H_1$**: $\mu_0 \neq \mu$ (meaning estimated our sample mean $\mu_0$ is different from the population mean value $\mu$)

### <a>Test Statistic</a>

We use the $z$ distribution (AKA, the standard normal distribution) to calculate the test statistic using the equation:

$$\color{dodgerblue}{\,\,\,\, \large z = \frac{\bar{x}-\mu}{\sigma / \sqrt{n}}}$$

<font color="dodgerblue">**This tells you how far your sample mean is from the known value, measured in standard errors (the denominator)**</font>. Just like with z-scores, a larger $z$ means your sample mean $\bar{x}$ is further from the population mean value $\mu$.

To convert this $z$ statistic to a $p$-value:
1. Find the probability of seeing that a sample mean that is **at least as extreme as yours**. You can do this two ways:
    - Use a $z$ distribution table to find the probability
    - Use the R function call <font color="tomato">`pnorm(abs(z), lower.tail = FALSE)`</font>
2. **Multiply the probability you get by two** to account for the two tails of the distribution.

### <a> Critical Value </a>

To find the threshold our test statistic must cross to reject $H_0$, we can:
- Use a $z$ distribution table for the correct $\alpha/2$ significance level, $z_{\alpha/2}$
- Use the <font color="tomato">`qnorm(<alpha/2>)`</font> R command.

### <a>Interpreting Results</a>

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/hypo_test_sum.png?raw=true" width=400/>
</figure>

Here, we have two options:
- Compare your $p$-value to your $\alpha$
- Compare your test statistic to your critical value

Both are valid and equivalent approaches since the critical value is the boundary that separates the <font color="tomato">rejection region</font> from the <font color="turquoise">non-rejection region</font>.

- If your <font color="tomato">**p-value $\leq \alpha$**</font>, it falls within the <font color="tomato">rejection region</font> and we <font color="tomato">**reject $H_0$**</font>.
    - This is also the case if <font color="tomato">**|test statistic| $\geq$ |critical value|**</font>.
    - We conclude that our sample mean $\mu_0$ is NOT the same as the population mean value $\mu$.
- If your <font color="turquoise">**p-value $> \alpha$**</font>, it falls within the <font color="turquoise">blue region</font> and we <font color="turquoise">**fail to reject $H_0$**</font>.
    - This is also the case if <font color="turquoise">**abs(test statistic) < abs(critical value)**</font>.
    - We cannot conclude anything about our sample mean $\mu_0$ in relation to the population mean value $\mu$.
    - Failing to reject $H_0$ does not prove the means are equal, it just means you do not have strong enough evidence to conclude that they are different.

<font color="mediumseagreen">Remember to write your interpretation within the proper context of the problem and to include your $p$-value.</font>

# <a> Two-Tailed 1-Sample Hypothesis $z$-Test Example </a>

----

A grocery store chain purchases a large batch of pre-made cakes from a large manufacturer. The manufacturer claims that their cakes weigh 500 grams with a standard deviation of 10 grams.

> The grocery store wants to know if the cakes they received comply with the manufacturer's claim, so they weigh each cake in their 250-unit order and note that their average cake weight is **498 grams**.

Following [the steps outlined before](#hypo_test_gen_proc):

### Step 1: Outline what we know and what we don't know
- The sample average weight $\bar{x}$ = 498 g
- The sample size $n$ = 250
- The manufacturer claimed average weight $\mu$ = 500 g
- The manufacturer claimed standard deviation $\sigma$ = 10 g

### Step 3: Choose the appropriate test
Since we are comparing our sample to a claimed population mean that we know, we have a sample size larger than 200, and know $\sigma$, we will use the two-tailed 1-sample $z$-test here.

### Step 3: Choose significance level
Let's use the standard $\alpha$ = 0.05 for this test. Since we are using a $z$ distribution, we need to find the z-score that coresponds to $\alpha$/2, which is $z_{\alpha/2} = z_{0.025}$.

We can either use the `qnorm(<alpha/2>)` function or look it up in a table to find that  $z_{0.025} = -1.96$.

### Step 4: Define your hypotheses
- $H_0$: $\mu_0$ = 500 g (we will assume this is correct and try to disprove it)
- $H_1$: $\mu_0 \neq$ 500 g

### Step 5: Calculate your test statistic

$\,\,\,\, \large z = \frac{\bar{x}-\mu}{\sigma/\sqrt{n}} = \frac{498-500}{10/\sqrt{250}} = \frac{-2}{0.63} = -3.16$

This means our sample mean $\bar{x}$ is -3.16 standard errors away from the population mean $\mu$.

### Step 6: Calculate the $p$-value from the test statistic

We want to know the probability of seeing values that are at least $\pm$3.16 standard deviations away from the mean, so we calculate $P(z \geq |-3.16|)$ using the R function `pnorm()`.

```
pnorm(abs(-3.16),lower.tail=FALSE)
```

We find that $P(z \geq |-3.16|) = 0.00078$

Therefore $p-value = 0.00078 * 2 = 0.00156$

### Step 7: Decide whether to reject $H_0$

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/hypo_test_ex2.png?raw=true" width=500/>
</figure>

Two ways to do this:
- Our p-value = 0.00156 is less than $\alpha$ = 0.05, so we reject $H_0$.
- The absolute value of our test statistic 3.16 is larger than the absolute value of our critical value 1.96, so we reject $H_0$.

### Step 8: Interpret the results in the context of our data

There is strong statistical evidence that the average cake weight in the order is different than the manufacturer claim of 500 g ($p = 0.0016$).

# <a> The two-tailed 1-sample $t$-test </a>
---

This test answers the same type of questions as the two-tailed 1-sample $z$-test we just learned about, but we use the $t$-test instead when...
- We have data from a relatively <font color="tomato">**small sample ($n \leq 200$)**</font> that is <font color="tomato">approximately normal</font>.
- We <font color="tomato">**do *NOT* know the population standard deviation $\sigma$, even if we have a larger $n$**</font>. Not knowing $\sigma$ means we have to approximate it using the sample, so we need to account for more uncertainty by using the $t$ distribution.
    - This is the case most of the time in real life scenarios, so the <font color="mediumseagreen">**$t$-test tends to be used a lot more than the $z$-test**</font>.

### <a>What we know</a>

- Your sample observations $x$
    - The sample mean $\bar{x}$
    - The sample standard deviation $s$
    - The sample size $n$
- The population mean value $\mu$

We can then calculate:
- The degrees of freedom for our data **$df = n - 1$**

### <a>What we don't know</a>

- The population standard deviation $\sigma$

The good news is that the hypotheses and how we interpret the results are all the same as the $z$-test! The main difference here is how we calculate the $p$-value.

### <a>Test Statistic</a>

We use the $t$ distribution to calculate the test statistic using the equation:

$$\color{dodgerblue}{\,\,\,\, \large t = \frac{\bar{x}-\mu}{s/\sqrt{n}}}$$

You might have noticed that the equation for $t$ is very similar to the equation for $z$, and it tells the same thing. What's new here other than using $\sigma$ instead of $s$ is how we convert this $t$ statistic to a $p$-value:
- Find the probability of seeing that a sample mean that is **at least as extreme as yours** by using the R function call <font color="tomato">`pt(abs(t), df, lower.tail=FALSE)`</font>.

3. You stil **multiply the probability by two** to account for the two tails of the distribution.

### <a> Critical Value </a>

To find the threshold our test statistic must cross to reject $H_0$, we can:
- Use a $t$ distribution table for the correct $\alpha/2$ significance level and degrees of freedom, $t_{\alpha/2, \, n-1}$
- Use the <font color="tomato">`qt(<alpha/2>, <n-1>)`</font> R command.
<br>

---
## Summary of Differences between the two-tailed 1-sample $z$ and $t$ tests

| Feature               | $z$-test           | $t$-test                       |
| --------------------- | ------------------ | ------------------------------ |
| Population SD (σ)     | Known              | Unknown                        |
| Distribution used     | Normal (Z)         | $t$ distribution               |
| Sample size           | $n \geq 200$ (or σ known) | Any (especially small $n \leq 200$)         |
| Variability estimate  | Exact (σ)          | Estimated (s)                  |
| Shape of distribution | Fixed              | Wider tails (more uncertainty) |

# <a> Two-Tailed 1-Sample Hypothesis $t$-Test Example </a>

----

A local clinical researcher wants to compare the Denver female type-2 diabetic population's blood glucose level to the national average type-2 female diabetic blood glucose level of 120 mg/dL.

> The researcher collects a **sample** of *45 adult female type 2 diabetic patients at UCHealth and measures their glucose level on a given day*. The sample mean $\bar{x}$ turns out to be **130 mg/dL** and the sample standard deviation $s$ turns out to be **30 mg/dL**.

Following [the steps outlined before](#hypo_test_gen_proc):

### Step 1: Outline what we know and what we don't know
- The sample mean blood glucose level $\bar{x}$ = 130 mg/dL
- The sample standard deviation $s$ = 30 mg/dL
- The sample size $n$ = 45
- The national average blood glucose level $\mu$ = 120 mg/dL
- The national standard deviation $\sigma$ = ???

### Step 2: Choose the appropriate test
Since we are comparing our sample to a known population mean that we know, we have a sample size smaller than 200, and do NOT know $\sigma$, we will use the two-tailed 1-sample $t$-test here.

### Step 3: Choose significance level
Let's use the standard $\alpha$ = 0.05 for this test. Since we are using a $t$ distribution, we need to find the $t$-statistic that coresponds to $\alpha$/2 and the degrees of freedom for our sample, which is $t_{\alpha/2, \, n-1} = t_{0.025, \, 44}$.

We can either use the `qt(<alpha/2>, <df>)` function or look it up in a table to find that  $t_{0.025, \, 44} = -2.02$.

### Step 4: Define your hypotheses
- $H_0$: $\mu_0$ = 120 mg/dL (we will assume this is correct and try to disprove it)
- $H_1$: $\mu_0 \neq$ 120 mg/dL

### Step 5: Calculate your test statistic

$\,\,\,\, \large t = \frac{\bar{x}-\mu}{s/\sqrt{n}} = \frac{130-120}{30/\sqrt{45}} = \frac{30}{2.12} = 2.24$

This means our sample mean $\bar{x}$ is 2.24 standard errors away from the population mean $\mu$.

### Step 6: Calculate the $p$-value from the test statistic

The degrees of freedom $df = n - 1 = 45 - 1 = 44$

We want to know the probability of seeing values that are at least 2.24 standard deviations away from the mean, so we calculate $P(t \geq |2.24|)$ using the R function `pt()`.

```
pt(2.24, 44, lower.tail=FALSE)
```

We find that $P(t \geq |2.24|) = 0.015$

Therefore $p-value = 0.015 * 2 = 0.03$

### Step 7: Decide whether to reject $H_0$

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/hypo_test_ex1.png?raw=true" width=500/>
</figure>

Two ways to do this:
- Our p-value = 0.03 is less than $\alpha$ = 0.05, so we reject $H_0$.
- The absolute value of our test statistic 2.24 is larger than the absolute value of our critical value 2.02, so we reject $H_0$.

### Step 8: Interpret the results in the context of our data

There is strong statistical evidence that the average blood glucose level of the sample of Denver female type-2 diabetic patients is different from the national average of 120 mg/dL ($p = 0.03$).

# <a> Doing a two-tailed 1-sample $t$-test in R with `t.test()`
---

You probably asked yourself "do we have to do all of this by hand every time?". Yes, to learn the concept... but no, not all the time!

R has a built-in function for $t$-tests: `t.test()`. It can run several types of t-tests, but to do a two-tailed 1-sample $t$-test like the ones we just did, follow these instructions:

1. Make sure your sample data is in a numeric vector (e.g., `dataframe$variable` or `x <- c(...)`).
2. Use the function call <font color="mediumseagreen">`t.test(<sample_data_vector>, mu=<population mean>, alternative="two.sided")`</font>.
    - By default, it will carry out the test with the default significance level of $\alpha = 0.05$. If you want a different significance level, add the parameter `conf.level=<1-alpha>` to your function call. You can also compare the $p$-value or $t$-statistic yourself.

This will return an object with a bunch of information about the test. The most useful to you might be:

- `$statistic` → the t-value
- `$parameter` → degrees of freedom
- `$p.value` → p-value
- `$estimate` → sample mean
- `$null.value` → hypothesized mean ($\mu_0$)
- `$alternative` → type of test (e.g., "two.sided")
- `$method` → description of the test

To specifically store and access these components, first make sure your test results are stored in a variable (e.g., `result`), then do `result$p.value`, `result$statistic`, etc.

# <a name="11appendixA">Appendix A: Why test the null hypothesis instead of the claim directly?</a>

---

In hypothesis testing, we initially assume the null hypothesis $H_0$ is true, then try to disprove it. We do this because **it's much easier to assess evidence against a claim than to definitively prove it true**.

- You can reject $H_0$ if evidence is strong.
- At the same time, you can never truly prove a hypothesis is correct, you can only fail to find evidence against it.

This is similar to a courtroom analogy. You may have heard the phrase "innocent until proven guilty".

> $H_0$ = “the defendant is not guilty", or "the defendant is innocent"

You look for evidence to *reject* that assumption - to say that "defendant IS guilty"

> $H_1$ = "the defendant is guilty"

- If your evidence is sufficient, you reject the $H_0$ claim (the defendant is innocent) in favour of the $H_1$ claim (the defendant is guilty).
- If evidence is insufficient, you *don't prove* the defendant is innocent, you just *fail to reject* that claim.

**Mathematically, we test $H_0$ because it gives us a clear baseline, we can calculate probabilities under it, and statistics is better at disproving than proving something.**

## Another way to look at it...

If you're familiar with debates or philosophy, this is a form of the *reductio ad absurdum* argument, which tries to discredit an idea by assuming that it is true, then showing that something contradicts it. Basically, **ff we can show that assuming $H_0$ is true makes something "absurd" happen, then we can reject the idea that $H_0$ is true.**
