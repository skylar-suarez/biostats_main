If the two groups have different spreads, it might not be fair to use a $t$ test to compare them. We will briefly cover what test(s) might be appropriate for that particular scenario next week. For now, we'll learn how to test for equal variances, then how to apply that method to compare the means of more than two groups.

# <a name="24two-q">Two-Tailed $F$-Test of Equal Variance</a>

---
<font color="mediumseagreen">We use an **$F$ test of equal variance** when we want to compare the variances of two independent normally-distributed groups.</font>

Some example questions we can answer with this test:
- "Is the systolic blood pressure of the control group more spread out than the experimental group's?"
- "Can I compare these two groups using a test that assumes equal variance?"

### <a> The Basic Idea </a>

> <font color="dodgerblue">We want to know whether the variances of two groups are significantly different from each other.</font>

To compare the variances of two groups, we calculate two variances and take the ratio. This ratio tells us how similar the variances are.
- The ***closer*** the ratio is to 1, the more ***similar*** the variances are.
- The ***further*** the ratio is to 1, the more evidence we have to say that the variances are ***different***.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/f_ratio_comp.png?raw=true" width=300/>
</figure>

*The $F$ test uses the $F$ distribution to tell you whether that observed ratio is so far away from 1 that it's unlikely to have occurred by random chance.*

### <a> When to use this test</a>

To use an $F$-test properly, we must have two approximately normal groups that are independent from each other.

### <a>What we know</a>

- The first set of measurements **$x_1$**
- The second set of measurements **$x_2$**

We can then calculate:
- The sample variance of first group **$s^2_1$**
- The sample varaince of second group **$s^2_2$**
- The degrees of freedom for the first group <font color="tomato">**$df_1 = n_1 - 1$**</font>
- The degrees of freedom for the second group <font color="tomato">**$df_2 = n_2 - 1$**</font>

### <a>Hypotheses</a>

Similar to means tests, we do not know the true variances of each population $\sigma^2_1$ and $\sigma^2_2$, but we can estimate them using our sample variances $s^2_1$ and $s^2_2$.

Knowing that, we write our hypotheses:

- **Null hypothesis $H_0$**: $\sigma^2_1 = \sigma^2_2$ (meaning our estimated variances are the same)
- **Alternative hypothesis $H_1$**: $\sigma^2_1 \neq \sigma^2_2$ (meaning our estimated variances are NOT the same)

### <a>Test Statistic</a>

We use the $F$ distribution to calculate the test statistic using the equation:

$$\color{dodgerblue}{\,\,\,\, \large F = \frac{larger\,\,variance}{smaller\,\,variance}}$$

- If the ratio is ~1, the variances are similar.
- If the ratio is far from 1, the variances are different.

You can convert this $F$ statistic to a $p$-value:
1. Find the probability of seeing that a difference in variance that is **at least as extreme as yours** by using the R function call <font color="tomato">`pf(F, df1, df2, lower.tail=FALSE)`</font>.
2. **Multiply the probability by two** to account for the two sides of the distribution.

### <a> Critical Value </a>

To find the threshold our test statistic must cross to reject $H_0$, we can:
- Use a $F$ distribution table for the correct $\alpha/2$ significance level and degrees of freedom, $F_{\alpha/2, \, df_1, df_2}$
- Use the <font color="tomato">`qf(<alpha/2>, <df1>, <df2>, lower.tail=FALSE)`</font> R command. **This will give you the upper boundary**.

#### *Note about this method*
Since the $F$ distribution is skewed right, the upper and lower boundaries are different.

- By forcing the larger variance into the numerator, we guarantee that the $F$ statistic will be *at least 1*. In this case, **we double the right-tail probability for our $p$-value.**
- If we didn't force the larger variance into the numerator, the $F$ statistic could be less than 1, and we would need to account for both tails differently.


### <a>Interpreting Results</a>


<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/f_test_summary.png?raw=true" width=400/>
</figure>

Comparing your p-value to your $\alpha$ is also very similar to previous tests:

- If your <font color="tomato">**p-value $\leq \alpha$**</font>, it falls within the <font color="tomato">rejection region</font> and we <font color="tomato">**reject $H_0$**</font>.
    - We conclude that our sample variances are not the same.
- If your <font color="turquoise">**p-value $> \alpha$**</font>, it falls within the <font color="turquoise">blue region</font> and we <font color="turquoise">**fail to reject $H_0$**</font>.
    - We cannot conclude anything about the similarity of our sample variances.
    - Failing to reject $H_0$ does not prove the variances are equal, it just means you do not have strong enough evidence to conclude that they aren't.

Just like before, remember to write your interpretation within the proper context of the problem and to include the $p$-value.

# <a> **Within-** & **Between-** Group Variances </a>
---

Understanding these two concepts will make the next test easier to understand. If we have several subgroups of data (e.g., different treatments, different disease risk categories, etc), there are two types of variances to consider.

## Within-Group Variance

The <font color="dodgerblue">**within-group variance**</font> of a dataset describes the variance within each group. We calculate this by doing a weighted sum of all the group variances and dividing that by the total sample size $N$ - the number of groups $k$:

$\large \frac{\sum_{i=1}^k (n_i-1)s^2_i}{N-k}$ where $N-k$ is its degrees of freedom.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/within_group_var.png?raw=true" width=700/>
</figure>

## Between-Group Variance

The <font color="dodgerblue">**between-group variance**</font> describes how different the means are from each other. To calculate it, we first do a weighted sum all the squared distances between each group's mean to the overall mean, then divide that by the number of groups $k$ - 1:  

$\large \frac{\sum_{i=1}^k n_i(\bar{y_i} - \bar{y})^2}{k-1}$ where $k-1$ is its degrees of freedom.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/between_group_var.png?raw=true" width=700/>
</figure>

## Why are talking about this?

Let's look at the possible combinations of these variances:

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/betweem_vs_within_group_var.png?raw=true" width=700/>
</figure>

We can see that it's much easier to tell the means are different when the within-group variance is low and teh within-group variance is high because there's less overlap between groups and the differences are more clear. This means that <font color="mediumseagreen">**The groups are more likely to have significantly different means when the between-group variance is much higher the within-group variance.**</font>

# <a name="24two-q">One-Way Analysis of Variance (ANOVA) Test</a>

---
<font color="mediumseagreen">We use a **One-Way ANOVA test** when we want to compare the means of three or more groups at once.</font> It is basically the multi-group equivalent of the two-tailed independent $t$ test. It is an extension of the $F$ test - in fact, it uses the same principle as the $F$ ratio combined with the Law of Total Variance to compare the means.

Some example questions we can answer with this test:
- "Do average blood pressures differ among patients across children, young adults, middle-aged, and senior?
- "Does weight lost differ accross three different diets?"

### <a> The Basic Idea </a>

> <font color="dodgerblue">We want to know whether at least one of the groups has a significantly different mean from the others.</font>

ANOVA uses the ratio of between-group variance to within-group variance to determine whether at least one group is significantly different from the others. Since it only cares when the ratio is LARGE, ANOVA is a <font color="dodgerblue">**right-tailed test**</font>. It's important to note that <font color="tomato">ANOVA does NOT tell you which groups differ from which others.</font> It only tells you whether there is evidence that not all means are equal.

### <a> When to use this test </a>

We use an ANOVA test when we have:
- 3+ groups of numerical observations
- The groups are approximately normal and have equal variances
- The observations in each group are independent of each other

### <a>What we know</a>

- The groups **$x_1$**, $x_2$, ... $x_k$ up to $k$ number of groups
- The sample size of all groups **$n_1$**, $n_2$, ... $n_k$
- The overall sample size **$N$**

We can then calculate:
- The mean of each group **$\bar{x_1}$**, **$\bar{x_2}$**, ... **$\bar{x_k}$**
- The overall mean of all observations, regardless of group **$\bar{X}$**  
- The variance of each group **$s^2_1$**, **$s^2_2$**, ... **$s^2_k$**
- The degrees of freedom of the between-group variance <font color="tomato">$df_b = k - 1$</font>
- The degrees of freedom of the within-gropu variance <font color="tomato">$df_w = N - k$</font>

### <a>Hypotheses</a>

Again, we do not know the true averages of each population $\mu_1$, $\mu_2$,..., but we can estimate them using our sample means $\bar{x_1}$, $\bar{x_2}$, ...

Knowing that, we write our hypotheses:

- **Null hypothesis $H_0$**: $\mu_1 = \mu_2 = ... \mu_k$ (meaning our estimated means are all equal)
- **Alternative hypothesis $H_1$**: AT LEAST ONE of our estimated means is different.

### <a>Test Statistic</a>

This is why we covered the variance types before! We use the $F$ distribution to calculate the test statistic using the ratio:

$$\color{dodgerblue}{\,\,\,\, \large F = \frac{between\,\,group\,\,variance}{within\,\,group\,\,variance} = \frac{\frac{\sum_{i=1}^k n_i(\bar{y_i} - \bar{y})^2}{k-1}}{\frac{\sum_{i=1}^k (n_i-1)s^2_i}{N-k}}}$$

Similar to the $F$ test,
- If the ratio is ~1, the between- and within-group variances are similar and therefore there is not much variation between the group means.
- If the ratio is >1, the between- and within-group variances are different, and therefore there is more variation between the group means.

You can convert this $F$ statistic to a $p$-value:
1. Find the probability of seeing that a difference in variance that is **at least as extreme as yours** by using the R function call <font color="tomato">`pf(F, dfb, dfw, lower.tail=FALSE)`</font>.
2. **You do NOT need to multiply the probability by two** since there is only one tail to account for.

### <a> Critical Value </a>

To find the threshold our test statistic must cross to reject $H_0$, we can:
- Use a $F$ distribution table for the correct $\alpha$ significance level and degrees of freedom, $F_{\alpha, \, df_b, df_w}$
- Use the <font color="tomato">`qf(<alpha>, <dfb>, <dfw>, lower.tail=FALSE)`</font> R command. **This will give you the upper boundary**. Since the $F$ distribution is skewed right, the upper and lower boundaries are different, but we only really need the upper boundary since the ratio will be $\geq$ 1.

### <a>Interpreting Results</a>


<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/anova_crit_val.png?raw=true" width=400/>
</figure>

Comparing your p-value to your $\alpha$ is also very similar to previous tests:

- If your <font color="tomato">**p-value $\leq \alpha$**</font>, it falls within the <font color="tomato">rejection region</font> and we <font color="tomato">**reject $H_0$**</font>.
    - We conclude that AT LEAST one of the groups has a different mean from the others.
      - note that we cannot tell which group is different.
- If your <font color="turquoise">**p-value $> \alpha$**</font>, it falls within the <font color="turquoise">blue region</font> and we <font color="turquoise">**fail to reject $H_0$**</font>.
    - We cannot conclude anything about the similarity of our group means.
    - Failing to reject $H_0$ does not prove the groups means are equal, it just means you do not have strong enough evidence to conclude that they aren't.

Just like before, remember to write your interpretation within the proper context of the problem and to include the $p$-value.


# <a> Doing a One-Way ANOVA Test `aov.test()`
---

I'm not going to ask you to do all that by hand or by raw code. However, we have to do a bit of prep work first.

1. Make sure your samples are in numeric vectors or a dataframe.
2. If your data is is contained in a numeric vector for each group, you need to combine them into a factor object (example below), then use the function call <font color="mediumseagreen">`aov.test(<observations> ~ <groupings>)`</font>.
2. If your data is in a dataframe that is separated by group, use the function call <font color="mediumseagreen">`aov.test(<observations> ~ <groupings>, data=<dataframe>)`</font>.

```r
groupA <- c(10.1, 9.8, 10.4, 9.9)
groupB <- c(11.2, 11.0, 10.8, 11.3, 11.1, 10.9)
groupC <- c(10.6, 10.7, 10.8, 10.5, 10.9, 10.4, 10.6)

observations <- c(groupA, groupB, groupC) # combines all observations together
groupings <- factor(c(rep("Group A", length(groupA)), rep("Group B", length(groupB)), rep("Group C", length(groupC)))) # creates an object that corresponds each entry to its group
groupings
```

It will return an object that you then read using the `summary()` function. An example is below.

```r
result <- aov(observations ~ groupings)
summary(result)
```

The `<groupings>` row is the important one here. The main things you want are the $F$ value and the $p$ value ("Pr(>F)" on the table). You can either compare the $F$ value to your critical value or compare the $p$-value to the significance level.

# <a name="hypo_test_gen_proc">Appendix: The General Process of Hypothesis Tests</a>
---

There are MANY different hypothesis tests, and in this class we will only cover a few of them. Still, most hypothesis tests follow the same general approach:

1. Outline what you know about your sample(s) and/or population.

2. Choose the appropriate test based on what you know and the question you want to answer.

3. Choose your <font color="dodgerblue">significance level</font>, the cutoff for "strong enough evidence to reject $H_0$", then convert it to a <font color="dodgerblue">critical value</font> for your test's distribution so you know where the threshold for rejection will be. The default is $\alpha$ = 0.05.

4. *Define the <font color="dodgerblue">null ($H_0$) and alternative ($H_1$) hypotheses</font>*. We will assume that $H_0$ is true and will try to disprove it.
    - $H_0$: "a drug has no effect", "two treatments are equally effective"...
    - $H_1$: "a drug does have an effect", "two treatments have different effect"...

5. *Calculate your <font color="dodgerblue">test statistic</font>*, a single number that measures how far your data is from what $H_0$ predicts. How we calculate this number differs depending on the specific test we're doing, but the idea is the same.

6. *Convert your test statistic to a <font color="dodgerblue">p-value</font>*, the probability of our data being at least this extreme if $H_0$ is true.
    -   The p-value is the area under both tails beyond the test statistic.
    -   The p-value is a probability, so we have $0 < \mbox{p-value} < 1$.
    -   The smaller the p-value, the stronger the evidence to reject $H_0$.

7. *Decide whether to reject $H_0$* or not by either comparing the p-value (our evidence of extremism) to the significance level (the pre-determined area of extremism), or comparing your test statistic (how far your data is from expected) to the critical value (the threshold for extremism).

    - If your <font color="tomato">**p-value $\leq \alpha$**</font>, it falls within the <font color="tomato">rejection region</font> and we <font color="tomato">**reject $H_0$**</font>.
        - This is also the case if <font color="tomato">**abs(test statistic) $\geq$ abs(critical value)**</font> - it means your data has passed the threshold for extremism and therefore have enough evidence to reject $H_0$.
    - If your <font color="turquoise">**p-value $> \alpha$**</font>, it falls within the <font color="turquoise">non-rejection region</font> and we <font color="turquoise">**fail to reject $H_0$**</font>.
        - This is also the case if <font color="turquoise">**abs(test statistic) < abs(critical value)**</font> - it means your data has not passed the threshold for extremism and therefore do not have enough evidence to reject $H_0$.

8. Interpret and summarize the results of our test within the context of the original question, including the $p$-value.
