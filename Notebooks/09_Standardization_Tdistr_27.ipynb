# <a name="08normal-pdf">The Probability Density Function for Normal Distributions</a>

---

The probability density function for a normal distribution
$X \sim N(\mu,\sigma)$ is called a <font color="dodgerblue">**Gaussian function**</font> whose formula is

$$\color{dodgerblue}{\large f(x; \mu, \sigma) = \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{1}{2} \left( \frac{x-\mu}{\sigma} \right)^2}}.$$

<font color="mediumseagreen">*Note: The formula for a Gaussian depends on both irrational numbers $\pi$ and $e$. How surprising and cool is that!*</font>



# <a name="08normal-area">Calculating Areas Under Normal Distributions</a>

---

If we had a normal random variable $X$ with $\mu$ = 26.5 and $\sigma$ = 4.5 and we wanted to find P(X < 18.5), we need to calculate the area to the left of $x$ under the normal curve with formula $f(18.5; 26.5, 4.5)$. We can try to calculate

$$P(X < 18.5) = \int_{0}^{18.5} \frac{1}{4.5\sqrt{2\pi}} e^{-\frac{1}{2} \left( \frac{x-26.5}{4.5} \right)^2} \, dx$$

<font color="tomato">*YIKES, good luck with that!* </font> We cannot find a formula for the
antiderivative of $f(x; 26.5, 4.5)$, so we need to find other methods
for calculating this area - thankfully, the hard work is already done for us if we convert our data to the **standard normal distribution**.



# <a name="08z-score">The Standard Normal Distribution</a>

---

The <font color="dodgerblue">**standard normal distribution**</font> is a special version of the normal distribution with a few fixed properties:

* Its **mean** (average) is **0**
* Its **standard deviation** is **1**

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/std_norm_dist.png?raw=true" width=400/>
</figure>


Its notation is $N(0,1)$.

### <a> Why it’s useful </a>

The standard normal distribution is useful because it lets you:
- Calculate probabilities much easier than using integrals
- Convert your data values into **z-scores**, which tell you how many standard deviations a value is from the mean
- Later, it will help us compare different data sets** on the same scale.


# <a name="08z-score">$z$-Scores</a>

---

A <font color="dodgerblue">**z-score**</font> is the <font color="dodgerblue">>number of standard deviations a value $x$ is away from the mean in a standard normal distribution</font>. You can calculate it using this equation:

$$\color{dodgerblue}{\large z = \frac{x-\mu}{\sigma}}$$

- The numerator of this equation is the distance between $x$ and the mean $\mu$.
- We then divide this distance by the standard deviation to give us the number of "steps" $x$ is away from the mean.

For example, if a value has a z-score of **+2.5**, it means the score is **2.5 standard deviations above the average**—which is relatively uncommon.

Doing this conversion with our data points is called <font color="dodgerblue">**standardization**</font> because it <font color="dodgerblue">converts our numbers (which could be any unit or magnitude) to the Standard Normal Distribution</font>. There are many advantages to this, as we will learn throughout the rest of the course.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/standardize.png?raw=true" width=600/>
</figure>

Going back to our cholesterol example from last week, the $z$-score of an adult with a total cholesterol level of 140 mg/dL is $z=\frac{140-200}{40} = -1.5$. This means a total cholesterol level of 140 mg/dL is 1.5 standard deviations *below* the mean mg/dL level.

<figure>
<img
src="https://upload.wikimedia.org/wikipedia/commons/a/a4/08fig-standardize.png"
data-fig-alt="Standardizing a normal distribution"
alt="Image Credit: Adam Spiegler, CC BY-SA 4.0, via Wikimedia Commons"  width="90%"/>
<figcaption aria-hidden="true">Image Credit: Adam Spiegler, <a
href="https://creativecommons.org/licenses/by-sa/4.0">CC BY-SA 4.0</a>,
via Wikimedia Commons</figcaption>
</figure>

- Comparing areas in the figure above, we see that $P(X<140) = P(Z < -1.5) $.
- <font color="dodgerblue">**Thus, using either the original distribution or the distribution of $z$-scores gives the same result.**</font>

### <font color="orange"> Key idea to remember </a>

<font color="orange">Your original data is likely not centered at 0 and doesn’t have a standard deviation of 1, but you can **transform it into the standard normal distribution** using a Z-score. Doing this makes calculating probabilities MUCH easier. That’s what makes it so powerful and widely used.<font>

## <a name="08normal-table">Using a z-score Table</a>

---

A z-score table tells us the <font color="mediumseagreen">**cummulative** area under the curve (AKA, the **cummulative probability**) of a standard normal distribution $N(0,1)$</font>, depending on how many standard deviations away we are from the mean (the z-score).

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/z_score_tbl.png?raw=true" width=800/>
</figure>

In the table above, we can see these probabilities for when we are *below* the mean (<font color="red">red</font>), and when we are *above* the mean (<font color="mediumseagreen">green</font>).

- For example, if $z$ = -0.85, we are 0.85 standard deviations *below* the mean. We would look at the *red* section of the table and locate the row where z is -0.8, then the column when z is 0.05, which gives us a probability of $\approx$ 0.198.
- Similarly, if $z$ = 0.85, we are 0.85 standard deviations *above* the mean. We would look at the *green* section of the table and locate the row where z is 0.8, then the column when z is 0.05, which gives us a probability of $\approx$ 0.802.

## <a name="08normal-R">Using R to Compute Areas</a>

---

Standard normal distribution tables were very useful prior to spread
statistical software and technology. If we are working in R, we do not
need to use tables to get approximations. Instead, we can use R
functions to compute these areas more accurately.

Similar to binomial, Poisson, and exponential distributions, R has built
in functions to perform calculations with normal distributions. In
particular, the function `pnorm()` calculates values of the cumulative
distribution function for a normal distribution:

- <font color="orange">`pnorm(x, mean, sd)`$=P(X<x)$ gives the area to the left of $x$ under $N(\mu,\sigma)$.</font>
- <font color="orange">`pnorm(x, mean,  sd, lower.tail=FALSE)`$=P(X>x)$ gives the area to the right of $x$ under $N(\mu,\sigma)$.</font>

Or you can find the $z$-score and use the standard normal distribution
in R as well:

-   `pnorm(z, 0, 1)`$=P(Z<z)$ gives the area to the left of $x$ under $N(0,1)$.
-   `pnorm(z, 0,  1, lower.tail=FALSE)`$=P(Z>z)$ gives the area to the right of $x$ under $N(0,1)$.

<font color="mediumseagreen">**Remember: Unlike discrete distributions: Do not use the function dnorm(x, mean, sd) to compute probabilities.** `dnorm()` returns the height of the probability density function, for continuous distributions probability is always that area under the distribution so you would always use `pnorm()`. </font>

# <a> Symmetry of the Normal Distribution</a>

---

Normal distributions are symmetric, which is a great advantage! It makes a lot of things easier - best explained in the image below:

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/norm_symm_props.png?raw=true" width=400/>
</figure>

So, if you only have one table, you can still calculate probabilities that are on the other side of the mean.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/neg_z_score_tbl.png?raw=true" width=800/>
</figure>

- If you need to calculate the probability of *"$z$ = 1.25 or greater"* (which would be in the <font color="mediumseagreen">positive</font> table) and you only have the <font color="red">negative </font> table, you can still use it to get that probability!
    - Since "$z$ = 1.25 or greater" is the same as "$z$ = -1.25 or less", you can find the former in the negative z-score table and see that it's $\approx$ 0.106.

- If you need to calculate the probability of *"up to $z$ = 1.18"* (which would be in the <font color="mediumseagreen">positive</font> table) and you only have the <font color="red">negative</font> table...
    - First, you would find the probability of up to $z$ = -1.18, which is $\approx$ 0.119.
    - Since "up to $z$ = -1.18" is the complement of "$z$ = 1.18 or less", you can do 1 - 0.119 to get $\approx$ 0.881.
    - If you scroll back up to the full tables, you can see that this agrees with its corresponding positive table value.

# <a>The Students' $t$ Distribution</a>
---

So far, we've learned that the bigger the sample size, the more closely our sample will resemble a normal distribution.

But what if you only have a small sample size and can't get any more samples?
<font color="tomato">**When your $n$ is less than 200, the normal distribution is not a faithful representation of the population.**</font>

This is where the <font color="dodgerblue">**$t$ distribution**</font> comes in - it is still bell-shaped and symmetrical, but it's wider and flatter, and its tails are thicker. All this *helps to account for the added uncertainty that comes from a smaller sample*.

Having a small sample size comes with some consequences - the fewer samples we have, the more "restricted" our data is. Basically, when you calculate something from you data (like the mean), the data points are no longer all completely independent of each other. In statistics, <font color="dodgerblue">**degrees of freedom (df)**</font> refer to <font color="dodgerblue">how many values in your dataset are still free to vary after you've used the data to estimate something about it</font>. For a bit more intuition behind this, check out the [appendix](#app).

The default calculation for the degrees of freedom is <font color="tomato">**$df = n - 1$$**</font>

- The smaller the sample size $n$ is, the wider and flatter the distribution is.
- The larger the sample size $n$ is, the more closely it resembles the normal distribution.

The plot below shows how the distribution shape changes as the number of samples increases.



```r
options(repr.plot.width = 10, repr.plot.height = 8) # adjust plot window size

x <- seq(-4, 4, length.out = 1000)

# Plot the standard normal distribution
plot(x, dnorm(x),
     type = "l",
     lwd = 6,
     col = "black",
     ylim = c(0, 0.42),
     xlab = "t",
     ylab = "Density",
     main = "t distributions with different df")

# Add t distributions
lines(x, dt(x, df = 3),  col = "darkgreen",       lwd =4)
lines(x, dt(x, df = 7),  col = "blue",      lwd = 4)
lines(x, dt(x, df = 30), col = "red",    lwd = 4)

legend("topright",
       legend = c("Normal", "df = 3", "df = 7", "df = 30"),
       col = c("black", "darkgreen", "blue", "red"),
       lwd = 4,
       bty = "n")
```

<font color="tomato">**NOTE: For now, we are not using the $t$ distribution to analyze data. Instead, we are focusing on what the $t$ distribution looks like and how it changes with sample size.**</font>

# <a name="app"> Appendix: Degrees of Freedom <a>
---

The concept of <font color="dodgerblue">**degrees of freedom (df)**</font> isn't very intuitive at first glance, but there is a way to make them easier to understand:

- Imagine you and a friend each roll a 6-sided die and want to try to guess what the other person rolled. You rolled a 3.
    - Right now, the value of what your friend rolled (Roll B) could be $\omega$ = {1, 2, 3, 4, 5, 6}.
    - Your probability of guessing correctly is 1/6.
    - This variable is free to vary - without any more information, it could be any of those numbers.
- Now... Imagine you have another friend look at both of your dice and they tell you that the sum of the rolls is either 8 or 9.
    - Knowing the sum imposes a constraint on what Roll B could be.
    - Since you know you rolled a 3 AND you know the potential sum totals, Roll B is no longer free to vary - you can calculate that it's either 5 or 6!

> <font color="orange">Degrees of freedom are the number of values that are still free to vary after you've used the data to estimate something. Every time we estimate a parameter (like the mean), we "use up" one degree of freedom because it poses a constraint on the data.</font>
