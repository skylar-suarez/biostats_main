# <a name="O5notation">Bayes' Theorem</a>

----

We've seen that that a conditional probability changes what subset of the sample space we care about based on making an assumption. In the case of P(A|B), we care about the probability that event A happens assuming that event B already happened.

$P(A|B) = \frac{P(A \cap B)}{P(B)}$

<img src="https://github.com/skylar-suarez/biostats_Main/blob/main/Images/a_given_b.png?raw=true" width=300>

Similarly, $P(B|A) = \frac{P(B \cap A)}{P(A)}$.

<img src="https://github.com/skylar-suarez/biostats_Main/blob/main/Images/b_given_a.png?raw=true" width=300>

Well, if we solve each of those for P(A $\cap$ B) and substitute those in our original equations, we get these very useful equations in return:

$P(A|B) = \frac{P(B|A)*P(A)}{P(B)}$

$P(B|A) = \frac{P(A|B)*P(B)}{P(A)}$

Why are these helpful? **All this means that P(A|B) and P(B|A) are related to each other.**

Sometimes it's easier to calculate or estimate P(A|B) than P(B|A), or the other way around. The important part is that **if we manage to get information about one of them, we also get information about the other**.

Bayes' Theorem is essentially a rule for how conditional probabilities should change when given new data - a way to update your beliefs when presented with new infomation. It is the basis for <font color="dodgerblue">**inference**: *the process of using evidence and reasoning to make educated guesses about something that isn't guaranteed or directly stated.*</font> That might sound complicated, but you actually already do this every single day, multiple times a day.
- You cannot say with 100% certainty that the bus will be late tomorrow morning, but since it's been ~2 minutes late most days for the last month, you may infer that it will likely happen again.
- You cannot be 100% sure that the milk you have in your fridge that is now one day past its sell-by date is fine to drink, but since you've drank 1-day-past-date milk several times before and been fine, you could infer that you'll be fine this time.

<font color="orange">Every time that you are making assumptions about the world around you based on previous experience, you are using inference.</font>

# <a>Screening Tests</a>

----

Medical screening is an essential part of disease prevention. Primary care screening tests are less invasive (e.g., mammograms or blood tests), but do not always have the best accuracy. In an ideal world, a test will always predict reality correctly 100% of the time. Unfortunately, perfect tests do not exist - there is simply too much variability in human biology to make that possible.

Typically, if a patient tests positive in primary care, they will undergo a secondary care test that is more invasive, but also more accurate (e.g., a biopsy). Understanding the statistics behind screening tests and how these tests are used in medical decisions is important for any biomedical engineer.  

<br>

###What can go wrong when testing for something?

What a test predicts can be different from reality in two main ways:

A <font color="mediumseagreen">**False Positive**</font> *(FP)* is when a **test result is positive** for a condition that a **patient does NOT actually have**.
- Example: a blood test resulting positive for diabetes when the patient does not actually have diabetes.

A <font color="mediumseagreen">**False Negative**</font> *(FN)* is when a a **test result is negative** for a condition that a **patient actually DOES have**.
- Example: a blood test resulting negative for diabetes when the patient does actually have diabetes.

### Fortunately, there are also the scenarios when the test does get it right:

- A <font color="mediumseagreen">**True Positive**</font> *(TP)* is when a test result is positive for a condition that a patient DOES have.

- A <font color="mediumseagreen">**True Negative**</font> *(TN)* is when a test result is negative for a condition that a patient does NOT have.

<br>

These concepts can be summarized in something called a <font color="dodgerblue">confusion matrix</font>.

<img src="https://raw.githubusercontent.com/gibsonea/Biostats/main/Images/confmatr.png" width=300>

## <a> Properties of Diagnostic Tests </a>

The two main properties of a diagnostic test are <font color="dodgerblue">**sensitivity**</font> and <font color="dodgerblue">**specificity**</font>.

- <font color="dodgerblue">**Sensitivity**</font> is the **probability that a patient will have a positive test result if they have the disease**.  We can describe this as a conditional probability: $P(T+|D+)$.
  - You can think of it as "the proportion of true positives out of all actual positive disease cases".
  - It simplifies to $\large \color{dodgerblue}{sensitivity = \frac{TP}{TP + FN}}$   
- <font color="dodgerblue">**Specificity**</font> is the **probability that a patient will have a negative result if they do not have the disease**.  It is described by: $P(T-|D-)$.
  - You can think of it as "the proportion of true neagatives out of all actual negative disease cases".
  - It simplifies to $\large \color{dodgerblue}{specificity = \frac{TN}{TN + FP}}$

<font color="orange"> **A good screening test will have both high sensitivity and high specificity.**</font>

These test properties can be measured by a clinical study using a sample of patients who are known to have the disease and a sample of healthy controls.

# <a name="O5notation">Predictive Values</a>

----

If a patient receives a positive test result from a screening test, it would be useful to know the probability that they actually DO have the disease. This probability is called the <font color="dodgerblue">**Positive Predictive Value** (PPV)</font>: $P(D+|T+)$.

Likewise, if they test negative, it would be useful to know what is the probability that they do not have the disease. This is the <font color="dodgerblue">**Negative Predictive Value** (NPV)</font>: $P(D-|T-)$.

In order to find these predictive values, we need not only the sensitivity and specificity of the test, but the disease prevalence, $P(D+)$. The equations relating these comes from Bayes' Rule:

<br>

$PPV = \frac{\text{Sensitivity} \cdot P(D+)}{\text{Sensitivity} \cdot P(D+) + (1 - \text{Specificity}) \cdot (1 - P(D+))}$

<br>

$NPV = \frac{\text{Specificity} \cdot (1-P(D+))}{\text{Specificity} \cdot (1-P(D+)) + (1-\text{Sensitivity}) \cdot P(D+) }$

<br>

The good news is that they simplify!

Since the $PPV = P(D+|T+)$, we can think of it as "the proportion of true positives out of all positive test results", which is:

$\large \color{dodgerblue}{PPV = \frac{TP}{TP + FP}}$

And since $NPV = P(T+|D+)$, we can think of it as "the proportion of true negatives out of all negative test results", which is:

$\large \color{dodgerblue}{NPV = \frac{TN}{TN + FN}}$

# <a name="O5notation">Frequentist vs Bayesian Inference</a>

----

The **probability of an event** is the *relative frequency of the event occuring over an indefinitely large number of trials*. For example, if you toss a fair coin and calculate the # of heads you saw /total # of tosses, it will approach 1/2 for many tosses (or "trials").

This definition forms the basis for the <font color="mediumseagreen">**frequentist**</font> method of inference, which is the main approach to statistical inference used in statistical practice.

**However**, Bayes had other ideas about inference. He didn't believe that probability should be defined by frequency - he considered it a theoretical concept that can never be realized in practice. Instead, Bayesians consider two types of probability: a *prior probability* and a *posterior probability*.

- The <font color="mediumseagreen">**prior probability**</font> of an event is the observer's best guess of the probability of that event without (or before) seeing any data related to that event.

- The <font color="mediumseagreen">**posterior probability**</font> of an event is the updated likelihood that an event will occur after collecting some empirical data. You're essentially integrating new information related to that event to your prior probability.

From the previous flu example, the probability that a random person from the county you live has the flu is given by the prevelence of flu or P(D+) = 0.15. This is the **prior probability** - your probability of having the flu is 0.15 since that's the information you currently have. If you are then tested it comes back positive, your probability of having the flu increases. This is the **posterior probability**.

# <a name="O5notation">Receiver Operator Characteristics (ROC) Curve</a>

----

We covered how a good screening test has both a high sensitivity and a high specificity. Unfortunately for us, *these two forces work against each other: when you increase one, the other decreases*. So, we're always dealing with the tradeoff: do we care more about correct detection, or correct rejection?

More importantly, how do we know if our test is good at either? This is where ROC curves come in.

Consider the first example concerning a non-invasive diagnostic test for pulmonary hypertension (PH). The test uses Echocardiographic Imaging (EI) to measure the pulmonary artery diameter $X_{d}$ which ranges from 0.5 to 4. A larger artery diameter means the person is at risk of having PH.

For clinical testing, we need to choose a diagnostic cutoff value for $X_{d}$, basically a value that we use as threshold to decide if the test result is positive or negative.

<img
src="https://github.com/skylar-suarez/biostats_Main/blob/main/Images/ROCexplanation.png?raw=true" width="700">

In the image above, the vertical lines represent different cutoff values. If you look carefully, you will notice that **changing the threshold changes the # of true positives (TP), true negatives (TN), false positives (FP), and false negatives (FN).** Seeing how these change based on a variety of cutoff values would be very helpful in deciding what that cutoff should be.

Instead of plotting many different distributions, we can create a <font color="mediumseagreen"> **Receiver Operator Characteristics** (ROC)</font> curve to get a better idea of the accuracy of the test. In a general sense, An **ROC curve is a visual way to see how well a test distinguishes between two options at different cutoff values** - the tradeoff between sensitivity and specificity.

To create a ROC curve, you calculate the test sensitivity and specificity at different cutoff values. Let's do an example with the non-invasive test for PH.

First, we set our cutoff value to be $X_{d}$ = 4. In this case, the cutoff is set so high that no one in the study tests positive and everyone tests negative.

<img
src="https://github.com/skylar-suarez/biostats_Main/blob/main/Images/ROCcurveXd4.png?raw=true" width="400">

Sensitivity is the number of true positives (i.e. both disease positive and test positive) divided by number of disease positives:

- Sensitivity = 0/(0+41) = 0

Specificity is the number of true negatives (i.e. both disease negative and test negative) divided by the number of disease negatives:

- Specificity = 12/(0+12) = 1

So, for cutoff of 4,
- Sensitivity = 0
- 1 - Specificity = 1 - 1 = 0

Our point on the ROC curve would be **(0,0)**.

<br>

Next, we can choose a cutoff value that is at the lowest value of $X_{d}$ so that everyone is test positive (i.e. cutoff of $X_{d}$ = 0)

Our confusion matrix would then be:

<img
src="https://github.com/skylar-suarez/biostats_Main/blob/main/Images/ROCcurveXd0.png?raw=true" width="400">

- Sensitivity = 41/(41+0) = 1
- Specificity = 0/(12+0) = 0

For cutoff of 0,
- Sensitivity = 1
- 1 - Specificity = 1 - 0 = 1

Our point on the ROC curve would be **(1,1)**.

<br>

We now have the our two extremes. Note that all ROC curves include the points **(0,0)** and **(1,1)**.

<br>

Figure A and B show the same distributions but with different threshold values (vertical lines), showing the tradeoff between sensitivity and specificity as you change the cutoff. These are the changes that the ROC curve is describing - but is there a way to summarize all that information better?

The <font color="mediumseagreen"> **Area Under the ROC Curve** (AUC) </font> is a useful property that can tell you more about whether your test is accurate. **It basically tells you that the test is accurate x% of the time.**

If the test is perfect, the area under the curve would be 1, or an accuracy of 100% - i.e. always correctly identifies whether the person has the disease or not.  If you have a test that has some false positives and false negatives, then the AUC is less than 1.
