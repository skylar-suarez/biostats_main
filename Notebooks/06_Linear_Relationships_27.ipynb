# <a name="25intro">Linear Relationships Between Two Continuous Variables</a>

---

In biomedical research, we are often trying to answer questions like *"is age related to blood pressure?"* or *"does cancerous cell division rate decrease as treatment dosage increases?"*. To answer these, we need a way to describe how two variables move together, both in strength and direction.
- Does all the data follow a predictable linear path?
- When variable one increases, does the other do the same?
    - If so, by how much? In which direction?
- How closely related are these variables?

<font color="dodgerblue">**Correlation**</font> and <font color="dodgerblue">**linear regression**</font> are two ways to describe how two variables are linearly-related. Correlation summarizes the strength and direction of the relationship, while linear regression goes a step further by fitting a line that can help explain or predict one variable from the other.

The simplest relationship two variables can have is a linear relationship. <font color="dodgerblue">**This means that as one variable changes, the other changes by about the same amount each time**</font> - it doesn't speed up, slow down, or bend in a curve.

> *The easiest example of this is if you work a job that pays you per hour.* Unless your employer breaks the law, this relationship is perfectly linear because for every 1 hour you work, you get paid the same constant amount of money.

In real life (and especially in biosciences), most relationships are not perfectly linear or depend on only one variable. Biology and real life simply have too much variability for data to fit perfectly onto a straight line. We'll revisit that concept later.

The code block below shows and example of a linear relationship alongside two non-linear ones.

``````r
options(repr.plot.width = 15, repr.plot.height = 5) # adjust plot window size

x <- seq(1, 10, length.out=20)

par(mfrow=c(1,3))

linear_y <- 2 * jitter(x, amount=0.5) + 1
exp_y <- exp(0.5 * jitter(x, amount=0.5))
log_y <- 3 * log(jitter(x, amount=0.5), 25)+ 2

plot(x, linear_y, main="linear", xlab="X", ylab="Y")
plot(x, exp_y, main="exponential", xlab="X", ylab="Y")
plot(x, log_y, main="logarithmic", xlab="X", ylab="Y")
``````

# <a name="corr_intro">Pearson's Correlation Coefficient $r$</a>

---

Correlation measures the strength and direction of a linear relationship.

<font color="dodgerblue">**Pearson's correlation coefficient ($r$)**</font> is a way to quantify and summarize how two variables move together in a straight-line pattern. It calculates both strength and direction of the relationship.

A good way to think about correlation is that it's a measure of how closely the points in a scatter plot cluster around an ***imaginary*** straight line:
- The closer the points lie along the line, the stronger the relationship is.
- The more "upward" the line points, the more positive the relationship is.
- If the points are scattered without a clear pattern, there is likely no relationship.



``````r
options(repr.plot.width = 15, repr.plot.height = 5) # adjust plot window size

x <- seq(1, 10, length.out=20)

par(mfrow=c(1,3))

strong_y <- -1 * jitter(x, amount=0.5) + 1
weak_y <- -1 * jitter(x, amount=3) + 1
none_y <- rnorm(length(x), mean=0, sd=0.5)

plot(x, strong_y, main="strong correlation", xlab="X", ylab="Y")
plot(x, weak_y, main="weak correlation", xlab="X", ylab="Y")
plot(x, none_y, main="no correlation", xlab="X", ylab="Y")
``````

## <a> How to calculate $r$ </a>

If you have two random variables $X$ and $Y$ whose individual data points correspond to $x_i$ and $y_i$, the Pearson's correlation coefficient is:

$$r = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum_{i=1}^{n}(x_i - \bar{x})^2}\sqrt{\sum_{i=1}^{n}(y_i - \bar{y})^2}} $$

What's happening here:
- In the **numerator**, we're **comparing how each individual $x$ and $y$ value differs from its mean**. If both are above their means or both are below their means, the product here is positive. If one is above and the other is below, the product is negative. This gives us the direction of the relationship.
- In the **denominator**, we are rescaling the numerator by the amount of spread in each variable, **squishing down the relationship to a unitless number between -1 and +1**. This makes comparing numbers from different datasets with different units possible.
- Together, this equation *combines the paired differences in $X$ and $Y$ into one number that measures the strength and direction of the linear relationship*.

## <a> How to Interpret $r$ </a>

$r$ values range from -1 to 1. The sign tells you the direction, while the number tells you the strength of the relationship.
- The <font color="mediumseagreen">**larger** |$r$|</font> is (closer to 1), the <font color="mediumseagreen">**stronger**</font> the relationship is.
- The <font color="tomato">**smaller** |$r$|</font> is (closer to 0), the <font color="tomato">**weaker**</font> the relationship is.
- If <font color="mediumseagreen">$r$ has a positive (+) sign</font>, the relationship is positive.
    - This means that the variables move in the <font color="mediumseagreen">**same direction**. $↑X \, ≈ \,\,\, ↑Y$ </font> (as $X$ increases, $Y$ also increases), and the other way around.
- If <font color="tomato">$r$ has a negative (-) sign</font>, the relationship is negative.
    - This means that the variables move in <font color="tomato">**opposite directions** $↑X \, ≈ \,\,\, ↓Y$ </font> (as $X$ increases, $Y$ decreases), and the other way around.
- If $r$ is near 0, there is no correlation - either the variables are not related, or their relationship is not linear.

## <a> Conclusions from $r$ </a>
---

As always with statistics, it's <font color="dodgerblue">**important to interpret your results within the context of your original question**.</font>

As an example:

> Imagine you work at a car dealership and have a list of your lot's car ages and their monetary value. You know there's a relationship between a car's age and its value, but you don't know how strong it is.

You run a correlation calculation and find out that for those two variables, $r = -0.75$.
- You can tell that the *negative sign means the variables move in opposite directions*. In our case, it means that *as age increases, value decreases*.
- You can also tell that the |$r$| of 0.75 means that the *relationship is strong*. In our case, it means that the relationship between age and value is strong ($r$ = 0.75).

So, a **complete interpretation** of this correlation calculation would be:
> <font color="mediumseagreen">*"The average relationship between a car's age and its monetary value is strongly negatively correlated ($r$ = 0.75)."*</font>

# <a> In correlation... Does it matter which variable is which? </a>
---

Short answer: no.

Long answer: Correlation is about the relationship between two variables as a pair. The correlation calculation only cares about the distances between each $x$ and $y$ value and their means. It measures how the paired values move together, without distinguishing either as input or output. If you switch $X$ and $Y$ in a correlation calculation, the resulting $r$ will be the same.

``````r
options(repr.plot.width = 10, repr.plot.height = 5) # adjust plot window size

x <- c(1, 2, 3, 4, 5)
y <- c(1.5, 2.3, 3.1, 4.6, 5.8)

corr1 <- cor(x, y, method="pearson")
corr2 <- cor(y, x, method="pearson")

par(mfrow=c(1,2))

plot(x, y, main=paste("X vs Y, ", round(corr1, 3)), xlab="X", ylab="Y")
plot(y, x, main=paste("Y vs X, ", round(corr2, 3)), xlab="Y", ylab="X")
``````

## <a> Pros and Cons of Correlation </a>

Correlation is useful because it gives a simple summary of how strongly two variables move together and in what direction.

<font color="mediumseagreen">The good:</font>
- It's easy to compute
- It's easy to interpret
- It's unitless, making it easier to compare variables with different scales
- It's helpful for quickly spotting whether a linear relationship might exist

<font color="tomato">The bad:</font>
- **It only captures linear relationships**.
    - This means it can miss other types of relationships that might exist. There could very well be an exponential or logarithmic relationship between $X$ and $Y$. *Getting a low $r$ doesn't mean that a relationship doesn't exist - it just means a linear one doesn't exist.*
- **It can be easily influenced by outliers**.
    - Since $r$ is calculated using the difference between each $x$ and $y$ pair from their means, one or two outliers can have a big impact on the $r$ value. In the example below, we have the same $X$ and $Y$ variables, but I replaced one of the middle values with an outlier, and you can see how $r$ was affected by that change.

``````r
options(repr.plot.width = 10, repr.plot.height = 5) # adjust plot window size

x <- seq(1, 10, length.out=20)

par(mfrow=c(1,2))

pos_strong_y <- 2 * jitter(x, amount=0.5) + 1
pos_strong_w_outlier_y <- pos_strong_y
pos_strong_w_outlier_y[10] <- 30

pos_strong_y_corr <- cor(x, pos_strong_y, method="pearson")
pos_strong_w_outlier_y_corr <- cor(x, pos_strong_w_outlier_y, method="pearson")

plot(x, pos_strong_y, main=round(pos_strong_y_corr, 3), xlab="X", ylab="Y")
plot(x, pos_strong_w_outlier_y, main=round(pos_strong_w_outlier_y_corr, 3), xlab="X", ylab="Y")
``````

# <a> Association, Correlation, and Causation </a>
---

These three terms get thrown around a lot in the news or in everyday conversations, but they mean very different things. It's important for us as biomedical scientists to be aware of what these terms mean, how they're different, and how to interpret them when reading scientific literature:

- <font color="dodgerblue">**Association**</font> is a general relationship between two variables that tend to vary together in some way.
- <font color="dodgerblue">**Correlation**</font> is a statistical measure of the strength and direction of a linear relationship between two variables
- <font color="dodgerblue">**Causation**</font> is when a change in one variable *directly* causes a change in another.

These concepts matter because they help us avoid going from "these two things are related" to "this thing caused the other" without sufficient evidence from robust experimentation. Unfortunately, this sort of thing happens often in the news and in everyday conversations.

*Causation is very difficult to prove*. You need many highly robust experiments that remove confounders and test variables directly to conclude a causal relationship.

On the other hand, correlation can happen by chance, due to confounding variables, or many other factors. You could evaluate the relationship of two variables that are completely unrelated in real life and still see a correlation - this phenomemnon is called a <font color="dodgerblue">*spurious correlation*</font>. There's [an entire website](https://www.tylervigen.com/spurious-correlations) dedicated to these, including ["google searches for 'how to annex texas' correlates with the number of phlebotomists in georgia"](https://www.tylervigen.com/spurious/correlation/1107_google-searches-for-how-to-annex-texas_correlates-with_the-number-of-phlebotomists-in-georgia) and ["popularity of the first name Kelsey correlates with NASA's budget appropriation"](https://www.tylervigen.com/spurious/correlation/1031_popularity-of-the-first-name-kelsey_correlates-with_nasas-budget-appropriation). Does this mean that the increase of babies named Kelsey caused the increase in NASA's budget appropriation? Most likely not, but hearing that those two things are correlated might make you think they might!

For this reason, statisticians always shout the mantra:

## <font color="TOMATO">**CORRELATION DOES NOT MEAN CAUSATION!**</font>

And now you will too.

# <a name="lin_reg_intro">Simple Linear Regression</a>

---

So far with correlation, we calculated a single-number summary of how one variable changes as another changes in a linear way. What correlation doesn't give us is the line itself - <font color="mediumseagreen">*the **average** change in $Y$ that comes from a change in $X$*</font>. This is useful because sometimes we have a new input value $x$ and want to predict what its corresponding output value $y$ might be.

Some examples:
- Can we predict systolic blood pressure from weight?
- Can we predict cell division rates from temperature?

With linear regression, our goal is not to summarize the strength of the relationship, but to <font color="dodgerblue">**create a line that best fits the relationship**</font> between an independent variable $X$ and dependent variable $Y$. We can then use this line to make predictions. This line is expressed as:

$$\hat{Y} = b_0 + b_1X$$

<font color="gray" size=2>If this looks familiar, it should! This is the equation for a straight line. You may have learned it as $y = mx + b$, but the concept is the same. </font>

### **$\hat{Y}$ is the <font color="dodgerblue">"line of best fit"</font> that best fits the relationship between $X$ and $Y$, which we can use to predict future or unknown values of $X$.**

<font color="mediumseagreen">This equation gives you a way to calculate an expected value $y_i$ for any input $x_i$.</font>

In this equation, we have the following coefficients:
- **$b_0$ is the *intercept*** of the line (the value of $Y$ when $X = 0$).
- **$b_1$ is the *slope*** of the line (how much change in $Y$ comes from a change in $X$). <font color="dodgerblue">**This value tells us that for each 1-unit increase in $X$, $Y$ is expected to increase by $b_1$ units, on average**</font>. Just like the correlation coefficient,
    - If the slope is positive, the variables move in the same direction.
    - If the slope is negative, the variables move in opposite directions.

<font color="tomato">**Our job is to find the coefficients $b_0$ and $b_1$ that make up the line.**</font>

> Note: In linear regression, variable assignment matters. If you switch the independent and dependent variables, the resulting lines may not (and in most cases won't) be the same!

### <a> Intuition Example </a>

Let's go back to the dealership example and that list of car ages (in years) and their monetary value. You want to come up with a line that expresses what happens with the car's dollar value (your $Y$) when a its age increases by a year (your $X$). You do your calculations and come up with an equation for the line of best fit that says:

$\,\,\,\,\,\,\, \hat{Y} = 30,000 - 2,000 * X$ <font color="dodgerblue">(the blue line in the plot below)</font>

Put into the context of your original question, this translates to:

$\,\,\,\,\,\,\, \hat{value} = 30,000 - 2,000 * age$

This makes our intercept $b_0 = 30,000$ and our $b_1 = -2000$. From this equation, we can deduce that, **on average**:
- Cars tend to start with a value of \\$30,000 at age = 0.
- For every year that a car ages, it **loses** \\$2,000 in value.

Now imagine a 10-year-old car comes to the lot and you're asked to assess its value. A good place to start would be that line of best fit equation:

$\,\,\,\,\,\,\, \hat{value} = 30,000 - 2,000 * 10 = 10,000$ <font color="tomato">(red dot below)</font>

You now have a predicted value of \\$10,000 for that car based on the average trend from the current cars on the lot.


``````r
options(repr.plot.width = 5, repr.plot.height = 5) # adjust plot window size

car_ages <- c(0.5, 1.2, 1.3, 1.5, 2.1, 2.8, 3.2, 3.7, 4, 4.8, 5.1, 6.4, 7.9, 8.4, 9.2, 9.5, 12, 14.2) # 18 cars in the lot
monetary_values <- 30000 - 2000 * jitter(car_ages, 50) # added a bit of noise to make it more realistic

plot(car_ages, monetary_values, main="Car Values based on Age", xlab="car age (years)", ylab="car value ($)")
abline(a=30000, b=-2000, col="blue") # y_hat, the line of best fit in the example (the parameter "a" is the intercept b_0, "b" is the slope b_1)

# 10-year-old car comes into the lot
car1_age <- 10
car1_value <- 30000 - 2000 * car1_age

points(car1_age, car1_value, col="green", pch=19)
``````

## <a> Getting the Equation for the Regression Line </a>
---

Our goal is to make the equation $\hat{Y} = b_0 + b_1X$

### Step 1: Calculating the slope $b_1$

If you calculate the Pearson correlation coefficient $r$, you've already done some of the work, you'll just need to calculate the standard deviations of $X$ and $Y$:

$$b_1 = r \cdot \frac{s_y}{s_x}$$

Alternatively, you can also calculate the slope using the data directly:

$$b_1 = \frac{\sum_{i=1}^{n}(x_i - \bar{x})(y_i - \bar{y})}{{\sum_{i=1}^{n}(x_i - \bar{x})^2}}$$

### Step 2: Calculating the intercept $b_0$

This is simply calculated using

$$b_0 = \bar{Y} - b_1 \bar{X}$$

Where $\bar{Y}$ is the mean of $Y$ and $\bar{X}$ is the mean of $X$.

### Step 3: Putting it together

Once you have your coefficients, you can plug them into the original equation from the beginning.



## <a> Step 4: Interpreting the Regression Line </a>
---

Once we've got the equation, the last step is to translate it to the specific language and context of our question. Remember to include units in this case!

The general structure for interpreting a linear regression line like this is:

> We found a <font color="magenta">[positive/negative]</font> linear relationship where for every <font color="magenta">[1-unit increase in the independent variable]</font>, the <font color="magenta">[dependent variable]</font> is expected to <font color="magenta">[increase/decrease]</font> by approximately <font color="magenta">[$b_1$] [units of the dependent variable]</font>.

In our car dealership example, we had an equation of $\hat{value} = 30,000 - 2,000 * age$

We can interpret this by saying:

> "We found a <font color="magenta">negative</font> linear relationship where for every <font color="magenta">year a car's age increases</font>, its <font color="magenta">monetary value</font> is expected to <font color="magenta">decrease</font> by approximately <font color="magenta">\\$2,000</font>."

### Some things to remember
- **Focus on "average"**. The regression line we've built represents a model for average trends, not a perfect prediction for every individual data point.
- **Direction matters**. If the slope is positive, the variables move in the same direction. If they are negative, the move in opposite directions.

# <a name="reg_line_example"> Full Example: Hours of Sunshine and Ice Cream Sales </a>
---

We have 5 days' worth of information from an ice cream truck:

| Hours of Sunshine | # of Ice Creams Sold |
|-|-|
|2 | 4|
| 3 | 5|
|5 | 6|
|7 | 10 |
| 9 | 15 |

Our job is to try to find a line that best fits the relationship between hours of sunshine (our $X$) and the numbrer of ice creams sold (our $Y$).

First, let's visualize our data:

``````r
hours_sunshine <- c(2, 3, 5, 7, 9)
num_ice_creams_sold <- c(4, 6, 7, 10, 15)

plot(hours_sunshine, num_ice_creams_sold)
``````

That looks pretty linear. Now let's come up with the equation:

### Step 1: Calculating the slope $b_1$

$b_1 = r \cdot \frac{s_y}{s_x}$

``````r
b1 <- cor(hours_sunshine, num_ice_creams_sold, method="pearson") * (sd(num_ice_creams_sold)/sd(hours_sunshine))
b1

# alternatively
sum((hours_sunshine-mean(hours_sunshine))*(num_ice_creams_sold-mean(num_ice_creams_sold))) / sum((hours_sunshine-mean(hours_sunshine))^2)
``````

This tells us that for every hour of sunshine in a given day, we can predict that the truck will sell ~1.45 ice creams.

### Step 2: Calculating the intercept $b_0$

$b_0 = \bar{Y} - b_1 \bar{X}$

``````r
b0 <- mean(num_ice_creams_sold) - b1*(mean(hours_sunshine))
b0
``````

### Step 3: Putting it together

The equation for a straight line $\hat{Y} = b_0 + b_1X$ in our context becomes:

$\,\,\,\,\,\,\,\,\,\,\hat{number\,of\,ice\,creams\,sold} = 0.85 + 1.45 * hours\,of\,sunshine$

We can see how the line sits on the data:



``````r
plot(hours_sunshine, num_ice_creams_sold)
abline(a=b0, b=b1, col="blue") # y_hat, the line of best fit (the parameter "a" is the intercept b_0, "b" is the slope b_1)
``````

### Step 4: Interpreting the regression line

We found a positive linear relationship where for every hour of sunshine on a given day, the truck's sales are expected to increase by approximately ~1.45 ice creams.

# <a> Residuals </a>
---

At the beginning of this notebook, I mentioned that almost nothing is perfectly linear, especially in biomedical sciences. Biology and real life simply have too much variability for data to fit perfectly onto a straight line.

To evaluate how well a straight line fits data, we first have to understand the concept of <font color="dodgerblue">**residuals**</font>.

Because real-world data is rarely ever perfectly linear, the actual data points will fall somewhere below or above the line. Put simply, a <font color="dodgerblue">**residual**</font> is the <font color="dodgerblue">difference between any observed $y_i$ value and the value predicted by our line $\hat{y_i}$</font>. It's also sometimes referred to as "error".

$$ res_i = y_i - \hat{y_i}$$

Just like everything else, these residuals have a direction:
- If the residual is **positive**, the **observation lies *above* the line**.
- If the residual is **negative**, the **observation lies *below* the line**.

<figure>
<img src="https://github.com/gibsonea/Biostats/blob/main/Images/residuals.png?raw=true" width=200/>
</figure>






## <a> Summarizing Error: Sum of Squared Residuals (SSR) </a>

Since residuals can be positive or negative, adding them together as it is would mean some might cancel each other out, making the total residual seem smaller than it actually is. To fix this, we square each residual:

$$res_i^2 = (y_i - \hat{y_i})^2$$

Graphically speaking, the squared residual is the area of a square that represents the error - the difference between the observed data point and the value predicted by the line of best fit.

<figure>
<img src="https://github.com/gibsonea/Biostats/blob/main/Images/reg_squares.png?raw=true" width=300/>
</figure>

To quantify these differences between predicted and observed, **we add all of the squared residuals together** to come up with the <font color="dodgerblue">**sum of squared residuals (SSR)**</font>. This number is a <font color="dodgerblue">how much variance in your data the line does NOT account for - the amount of "error" in your model</font>, *in the original units of your dependent variable*.

$$ \sum_{i=1}^{n} (y_i - \hat{y_i})^2 $$

> Note: the process we used to find the unique slope and intercept of the line *minimizes the area of these residual squares*. Because of that, this method is usually called the <font color="dodgerblue">**method of least squares**</font>.

## Calculate the SSR using R

The object `lm()` returns also contains the residuals, already calculated for you. You can get them by appending `$residuals`, just like you did for the coefficients. All you have to do is square and sum them.

`````
result = lm(y ~ x)
SSR = sum(result$residuals^2)
`````

## Interpreting the SSR

It's hard to interpret the SSR on its own. If you see an SSR of "500", we cannot definitively say whether it's "good" or "bad" without context. The context here might be the units (e.g., "500 grams" vs "500 kilograms"), and whether that's allowable based on your knowledge of the problem.

Something else to consider is that since we're just adding the squared residuals together, <font color="tomato">**the more data points you have, the larger the SSR will be**</font>.

But remember: the SSR tells you how much discrepancy there is between your observed data points and your predicted points!

# <a> How well the line fits the data: Coefficient of Determination $R^2$ <a/>
---

The SSR tells us the amount of error, but it doesn't actually measure how well the line we came up with fits the data. We need a number that actually quantifies that and is also unitless so we don't need to worry about whether the variables are on different scales (e.g., one ranges between 0 - 10 and the other ranges between 100 and 1000).

That number is the <font color="dodgerblue">**coefficient of determination $R^2$**</font>: a unitless quantitative measure of how much variance in the dependent variable is explained by the line we came up with.

For simple linear regression, we get that $R^2$ value by just squaring the correlation coefficient:

$$R^2 = r^2$$

This works because $r$ is already built from the variances of your variables. We lose the direction (the sign) by squaring it, but we link the strength of the relationship to the proportion of variation the regression line doesn't account for.

## Interpreting $R^2$

The $R^2$ ranges between 0 and 1:
- The number is the <font color="mediumseagreen">**proportion of variance in your dependent variable that is explained by your line**</font> (e.g., a line with an $R^2$ of 0.74 explains 74% of the variance in cars' monetary value)
- <font color="mediumseagreen">**The higher the number, the better the line fits the data.**</font>

## Calculating $R^2$ after using `lm()`

The object `lm()` returns doesn't have the $R^2$ value already calculated, but we can call `summary()` of your results and append `$r.quared` to it.

Example:
````
result = lm(y ~ x)
summary(result)$r.squared
````
