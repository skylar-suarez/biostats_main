## <a> Approximating the population mean $\mu$ with a sample </a>

---

In real life, *it's almost always impossible to measure an entire population* (too expensive, too slow, etc), so <font color="tomato">**we have no real way of knowing for sure what its mean is**.</font>

Yet, most scientific or biomedical questions have something to do with what the population average is (average blood pressure, average glucose level, etc).

- Imagine you want to know the average blood glucose level of all female type-2 diabetes patients in the Denver area. It's practically impossible to measure every single one of those people - we (or they) might not even know who they all are!
    - But what we *can* do is take a sample of 200 of them from our hospital and measure their glucose.
- Now, imagine that the normal distribution represents all possible sample means if you were to repeat your 200-person study many, many times.
    - This means that the "true" population mean lies somewhere within this distribution. But where? Is there a range of values it is more likely to fall within? How sure can we be of this?

This is where <font color="dodgerblue"> **confidence intervals** </font> come in. They help us approximate the "true" population mean using a subset of that population by *quantifying how confident we are that the true population mean $\mu$ lies within a range of values centered around our sample mean $\bar{x}$*.

<br>

# <a name="19unknown">What is Uncertain and What is Unknown?</a>

---

It can be easy to confuse the terms *uncertain* and *unknown*. In a less
formal setting, people may use these terms interchangeably. In
statistical inference, there is a clear distinction in their meaning
that is important to clarify:

-   Each  ***sample statistic* is a *known* value that is<font color="dodgerblue"> uncertain**</font> due to sampling.
    -   Each sample has its own measures - we *know* what each sample's mean $\bar{x}$ and standard deviation $s$ are without speculation.
-   The  ***population parameter* is an <font color="dodgerblue">unknown</font> value that is certain** (not variable).
    -   The parameter we are trying to estimate (in this case, $\mu$) is a fixed (certain) value, even if it's unknown.

Each study done by every scientist has a different estimate, but are trying to hit the same fixed target.

We cannot be certain whether our estimate is successful or not if the population mean $\mu$ is unknown!

<br>

# <a name="19intro">Confidence Intervals (CI)</a>

---

At its core, a confidence interval is a way to say "this is my best guess of what the population mean is, and how uncertain that guess is". Formally, it's a range of plausible values for a population parameter (in this case, the mean).

- This can help us answer questions like "what is the prevalence of type 2 diabetes in female people in the Denver area?", or "How many kids get the flu in Colorado elementary schools in the month of February?".

We create confidence intervals by taking a sample from a population, measuring what we care about, taking the mean ($\bar{x}$), and use that mean to estimate how likely it is that we have captured that unknown population mean ($\mu$).

The general idea is illustrated here:

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/CI_explain.png?raw=true" width=300/>
</figure>

Maths-wise, it will look like this:

$$ CI = \bar{x} \pm margin\,\, of\,\, error $$

Notice that a confidence interval has two major parts:
- To the *left* of the $\pm$: a central quantity (your sample mean $\bar{x}$)
- To right *right* of the $\pm$: a margin of error that sets up the lower and upper limits of our interval (we'll get into this in a bit)

This is equivalent to an interval. For example:

$CI = 5 \pm 3 = [2, 8] $  or   $2 \leq \mu \leq 8$

Right now, we will focus on calculating <font color="dodgerblue">**parametric**</font> CIs, which assume that the data comes from a specific underlying distribution. <font color="mediumseagreen">**We will start with using the normal distribution first.**</font>

# <a name="19intro"> Confidence Levels and $\alpha$ values </a>

---

The <font color="dodgerblue">**confidence level**</font> is the <font color="dodgerblue">probability that the interval **CONTAINS the true population mean $\mu$**</font>. It describes how reliable that interval building procedure is in the long run.

The <font color="dodgerblue">**significance level ($\alpha$)**</font> is the <font color="dodgerblue">probability that the interval **FAILS TO CONTAIN the true population mean $\mu$**</font>. It's defined as 1 - confidence level.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/CI_comparison.png?raw=true" width=600/>
<figcaption> (a) shows a 90% confidence interval, (b) shows a 95% confidence interval
</figcaption>
</figure>

In the image above, the coloured areas represent the confidence interval that correspond to each confidence level.
- Since our sample mean is the center of these curves, **the interval is built from the center outward**.
- If we want to be *more confident* that we've captured $\mu$, we must *include more possible values*, so the *interval gets wider*.

Conversely, the blank "tails" on either side represent $\alpha$ - the areas of the curve NOT included in the confidence interval, where $\mu$ is not likely to be.

**We use these levels to find the right <font color="dodgerblue">critical values</font> to build our confidence intervals.** To get a specific confidence level interval, we want half of the alpha to be on the left tail and the other half on the right tail, just like the image above shows.

## Critical Values

A critical value is simply a number that tells you "how many standard error wide" your confidence interval should be. In the *standard normal distribution*, it is represented by a <font color="dodgerblue">**z-score**</font>, which you learned about last week (the number of standard deviations away from the mean of a normal distribution for a given value).

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/CI_alpha_graph.png?raw=true" width=600/>
</figure>

A critical value is typically expressed as $z_{\alpha/2}$. Since our confidence interval is centered around the sample mean, The $\alpha/2$ corresponds to the area in each tail of the normal distribution.

> Important note: Since there are two tails, technically speaking, there will be two critical values: a *negative* one (left of the mean) and a *positive* one (right of the mean). They will have the same value.

So, for a confidence level of 80%:
- $\alpha$ = 0.2
- each tail is $\alpha/2$ = 0.1
- So we need the critical value $Z_{0.2/2} = Z_{0.1}$

We find these critical values on a table or using computational software. To find them on a z-score table, find the probability you want to cover on the left side, then make note of the the z-score that corresponds to that.

- For our 80% confidence level example, the table below tells me that the critical value $Z_{0.1}$ is 1.28.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/neg_z_score_tbl.png?raw=true" width=800/>
</figure>

To save you time, here's a table that summarizes the most common critical vlaues from a standard normal distribution for confidence intervals:

| Confidence Level | α (alpha) | Critical Value (z) |
| ---------------- | --------- | ------------------- |
| 90%              | 0.10      | 1.645               |
| 95%              | 0.05      | 1.96                |
| 98%              | 0.02      | 2.33                |
| 99%              | 0.01      | 2.575               |


<font color="dodgerblue">**The most common confidence level is 95% (or 0.95), which corresponds to $\alpha = 0.05$.** and its critical value $Z_{0.025}$ is 1.96 </font> This balances beting reasonably sure without making intervals too wide to be useful.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/95_CI_graph.png?raw=true" width=600/>
</figure>

## Putting it all together

So, if you calculate a CI with an alpha of 0.05, it means that
- If you repeat the entire study (sample from the population, calculate the mean, build the confidence interval) 100 times:
    - ~95 of those intervals would capture the true population mean $\mu$
    - The other ~5 of the intervals would miss it.
- <font color="dodgerblue">**The 95% is a long‑run success rate for the *procedure***</font>, not a probability about the single interval you happen to have (we will cover more on this later).

# <a name="19intro"> Equation for a Confidence Interval (normal distribution)</a>

------
The equation for a CI that approximates a normal population mean from a sample is:

$$ \large CI = \bar{x} \pm {z_{\alpha/2}} \cdot \frac{s}{\sqrt{n}}$$

Where:
- $\bar{x}$ is your sample mean
- $z_{\alpha/2}$ is the critical value that corresponds to your $significance/2$ to represent both tails of the distribution
- $s$ is the standard deviation of your sample
- $n$ is your sample size

The $\frac{s}{\sqrt{n}}$ term is called the <font color="dodgerblue">**sample standard error (SE)**</font>. It tells us how much a sample mean is expected to vary from sample to sample when we repeatedly take samples from the same population.

# <a name="19intro"> Confidence Interval Example (normal distribution) </a>

----

> You want to estimate the *average glucose level of adult type 2 diabetic female patients in the Denver area* **with 95% confidence**.

> To approximate this population mean, you take a **sample** of *200 adult female type 2 diabetic patients at UCHealth and measure their glucose level on a given day*.

### **Step 1**: Take your 200-patient sample and calculate:
- the sample mean $\bar{x}$
- the sample standard deviation $s$

> For this example, your sample mean $\bar{x}$ turns out to be **150 mg/dL** and your sample standard deviation $s$ turns out to be **30 mg/dL**.

### **Step 2**: Calculate your standard error (SE):

$\,\,\,\, \large \frac{s}{\sqrt{n}} = \frac{30}{\sqrt{200}} = 2.12$

### **Step 3**: Look up the critical value that corresponds to your desired significance level:
1. For a 95% confidence level, your significance level $\alpha$ will be 0.05. This means we need the z-score that puts 0.025 on each tail of the distribution
2. Look at a z-score table or use R to find the critical value that corresponds to that, which is 1.96.

### **Step 4**: Calculate your CI

$\,\,\,\, \large CI = \bar{x} \pm {z_{\alpha/2}} \cdot \frac{s}{\sqrt{n}}$

$\,\,\,\, \large \,\,\,\,\,\,\,\,\,\,\,\,= 150 \pm 1.96 \cdot 2.12$

$\,\,\,\, \large \,\,\,\,\,\,\,\,\,\,\,\,= 150 \pm 4.155$

### **Step 5**: Convert your CI to an interval for better interpretability

$\,\,\,\, CI = 150 \pm 4.115 = [145.8,154.2]$

### **Step 6**: Interpret your CI in the context of your question, including units

There are two widely-used ways to write out CI results:

- "Using a 95% confidence interval, the true average glucose level for adult type 2 diabetic women in the Denver area is estimated to be between 145.8 and 154.2 mg/dL $(n=200)$."
- "We are 95% confident that the true average glucose level for adult type 2 diabetic women in the Denver area lies between 145.8 and 154.2 mg/dL $(n=200)$."

Both are acceptable, but they do lack the nuance and intricacies of what CIs really mean (more on that in the misconceptions section ahead).

# <a>CIs for Smaller Sample Sizes</a>
---

Last week, we learned some intuition behind what the $t$ distribution represents and how it's useful for smaller sample sizes.


<font color="mediumseagreen">**If our sample size $n$ is smaller than 200, we should use the $t$ distribution to estimate a population mean.**</font>


Similar to the normal distribution, it has a very complicated PDF equation that we do NOT want to try to integrate! The good news is that it also has a handy table of <font color="dodgerblue">**$t$-values**</font> that work just like z-scores.

## Using a $t$ values table

There a couple of small differences between a $z$ table and a $t$ table.

First, there is a new term called <font color="dodgerblue">**degrees of freedom**</font>. It's basically describing how much information you have left after estimating something (in this case, the mean).

$$ df = n -1 $$

Unlike a $z$ table, a $t$ table gives you critical values based on different tail probabilities and degrees of freedom.

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/t_dist_tbl.png?raw=true" width=500/>
</figure>

## Finding the $t$-value using R

The function `qt(p, df, lower.tail=FALSE)` gives the critical value for the two-tailed alpha level $p$ (make sure you divide by 2!) and degrees of freedom $df$.

For example, to find the $t$-value for an 80% confidence interval for a sample of 20 data points:

- The significance level $\alpha$ is 1 - 08 = 0.2
- The degrees of freedom is 20 - 1 = 19

So we use `qt(p=0.2/2, df=19, lower.tail=FALSE)`

# <a>Calculating a CI using the Students' $t$ Distribution</a>
---

Even more good news: calculating a CI with the $t$ distribution is the same process as we just learned, but using a $t$-value instead.

$$ \large CI = \bar{x} \pm {t_{\alpha/2,\, df}} \cdot \frac{s}{\sqrt{n}}$$

> You want to estimate the *average glucose level of adult type 2 diabetic female patients in the Denver area* **with 95% confidence**.

> To approximate this population mean, you take a **sample** of *<font color="mediumseagreen">**20**</font> adult female type 2 diabetic patients at UCHealth and measure their glucose level on a given day*.

### **Step 1**: Take your 20-patient sample and calculate:
- the sample mean $\bar{x}$
- the sample standard deviation $s$

> For this example, your sample mean $\bar{x}$ turns out to be **150 mg/dL** and your sample standard deviation $s$ turns out to be **30 mg/dL**.

### **Step 2**: Calculate your standard error (SE):

$\,\,\,\, \large \frac{s}{\sqrt{n}} = \frac{30}{\sqrt{20}} = 6.71$

### **Step 3**: Calculate your degrees of freedom (df):

$\,\,\,\, df = n -1 = 20 - 1 = 19$

### **Step 4**: Look up the critical value that corresponds to your desired significance level:
1. For a 95% confidence level, your significance level $\alpha$ will be 0.05. This means we need the t-score that puts 0.025 on each tail of the distribution
2. Look at a z-score table or use R to find the critical value that corresponds to that with df = 19, which is 2.093.

### **Step 5**: Calculate your CI

$\,\,\,\, \large CI = \bar{x} \pm {t_{\alpha/2\, df}} \cdot \frac{s}{\sqrt{n}}$

$\,\,\,\, \large \,\,\,\,\,\,\,\,\,\,\,\,= 150 \pm 2.093 \cdot 6.71$

$\,\,\,\, \large \,\,\,\,\,\,\,\,\,\,\,\,= 150 \pm 14.04$

### **Step 6**: Convert your CI to an interval for better interpretability

$\,\,\,\, CI = 150 \pm 14.04 = [135, 164]$

### **Step 7**: Interpret your CI in the context of your question, including units

There are two widely-used ways to write out CI results:

- "Using a 95% confidence interval, the true average glucose level for adult type 2 diabetic women in the Denver area is estimated to be between 135 and 164 mg/dL  $(n=20)$."
- "We are 95% confident that the true average glucose level for adult type 2 diabetic women in the Denver area lies between 135 and 164 mg/dL $(n=20)$."

# <a> Using Confidence Intervals </a>
---

At the beginning of this lesson, we covered how a CI is a range of plausible values for a population parameter (in this case, the mean).

## Comparing to a Specific Value

One of the ways we can use CIs is to compare the plausible values to some other value. Imagine we have a 95% CI for the true average sales of a local ice cream truck called "Icy Dream" (<font color="orange">$CI_1 = [11,19]$</font>). Let's say that the <font color="turquoise">local average</font> ice cream truck sales is ~18 (<font color="turquoise">A</font>), and the <font color="tomato">national average</font> ice cream truck sales is ~10 (<font color="tomato">B</font>).

<figure>
<img src="https://github.com/skylar-suarez/biostats_main/blob/main/Images/CI_use_1.png?raw=true" width=500/>
</figure>

- Does Icy Dream, on average, sell more than the local average?
  - Since <font color="turquoise">A</font> is within <font color="orange">$CI_1$</font>, it is a plausible value for Icy Dream's true average sales. Therefore, *Icy Dream does **not** seem to sell more than the local average*.
- Does Icy Dream, on average, sell more than the national average?
  - Since <font color="tomato">B</font> is **not** within <font color="orange">$CI_1$</font>, it not a plausible value for Icy Dream's true average sales - in fact, B is lower than the CI bounds. Therefore, *Icy Dream **does** seem to sell more than the national average.*

## Comparing to Another CI

Now let's imagine with have a 95% CI for another local ice cream truck, "Tasty Treat" (<font color="mediumseagreen">$CI_2 = [17,24]$</font>).

  - Notice that <font color="orange">$CI_1$</font> and <font color="mediumseagreen">$CI_2$</font> overlap between 17 and 19. This means that they share plausible values for true average sales! Without any further analysis*, *we cannot say that Tasty Treat and Icy Dream have different average sales*.

# <a name="10misconc">Common Misconceptions about CIs</a>

---

Here are some very common mistakes students (and many researchers) make when interpreting confidence intervals:

### #1: Treating the interval as a probability about the true mean $\mu$

<font color="red">**Mistake:**</font> Saying *“There is a 95% probability that the true mean is between 9.4 and 10.4 mg/dL.”*

> *Why this is a problem:* In this framework, the true mean μ is a fixed (but unknown) value; it does not move around. What changes from sample to sample is the interval. It either contains $\mu$ or it doesn't.

<font color="mediumseagreen">**Better:**</font> *“This 95% confidence interval was created by a method that, in the long run, produces intervals that contain the true mean 95% of the time.”*

<br>

### #2: Assuming a single interval is "correct"

<font color="red">**Mistake:**</font> Thinking that *“a 95% confidence interval means this particular interval is almost certainly correct.”*

> *Why this is a problem:* Different samples → different intervals. If you re-sample and re-do the 95% CI method, it will still miss the true mean $\mu$ 5% of the time, even when all assumptions are satisfied. The 95% refers to how well the method works in the long run across many hypothetical studies, not a guarantee that this one specific interval is right.

<font color="mediumseagreen">**Better:**</font> Emphasize that the 95% refers to what would happen if you could repeat the whole study many times, not a guarantee for this one study.

<br>

### #3: Applying the interval to individuals

<font color="red">**Mistake:**</font> Thinking that *“because the 95% confidence interval for the mean glucose is from 9.6 to 10.4 mg/dL, 95% of individual patients have glucose levels between 9.6 and 10.4 mg/dL.”*

> *Why it's a problem:* A confidence interval for the mean is about the **population** average, not about individual values. Individual values can be outside of the interval.

<font color="mediumseagreen">**Better:**</font> *“This 95% confidence interval gives a plausible range for the population mean glucose level. It does not tell us what proportion of individual patients fall inside that range.”*
